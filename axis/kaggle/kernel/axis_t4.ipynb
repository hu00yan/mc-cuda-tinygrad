```python
# ---- 7. the width sweep: the scan rung across mixin/op.py's strategy threshold
#
# THIS CELL IS THE FALSIFIER, and it is the reason this run exists rather than a
# decoration on it. The note's central claim is that `scan`'s 26x is a STRATEGY
# problem and not a capability or a scheduling one, and the strategy it names
# (`Tensor.cumsum` -> `_split_cumalu`, `SPLIT = 256`) switches decomposition at
# N = SPLIT*2 = 512:
#
#     if not isinstance(s:=self.shape[axis], int) or s <= SPLIT*2: return self._cumalu(axis, op)
#
# So the prediction is sharp and was written down BEFORE the sweep ran: the
# framework's nanoseconds per ELEMENT step up across N=512, and the
# hand-written reference's does not, because the operator's difficulty does not
# change at 512 and only the framework's strategy does.
#
# ROWS x N is held constant at 2^22 elements, so every point moves the same
# amount of data and the only thing varying is the decomposition. A smooth
# reference beside a stepped framework is the whole experiment; a smooth
# framework refutes the claim, which is why the verdict is computed in
# `_print_sweep` rather than narrated here.
#
# THE N LIST IS 1024 AND UP, AND VERSION 20 PROVED THAT MATTERS. A first attempt
# swept 64..8192 and found a 3.6x denominator swing across the threshold --
# which read as the predicted discontinuity. It was an ARTEFACT, and the harness
# found it: the sweep's own reference selector chose `axis_scan_v4` for N=4096
# and `axis_scan_v1` for N=1024, so the DENOMINATOR changed kernel at exactly the
# point the experiment was looking for a change in the NUMERATOR. Selection by
# correctness alone, PT = N/1024, is wrong for a timing; the fewest PT that
# covers the span is the fastest correct kernel, and that is what is selected now
# (`_ref_scan_symbol_for_span`, separate from the correctness gate
# `_scan_symbol` because they answer different questions). The sweep also cannot
# straddle N=512 on the reference side at all: no `axis_scan_*` kernel is
# CORRECT at N=384, 512 or 513, so those rows are UNAVAILABLE and correctly so.
# Sweeping only the range where a correct reference exists, and pinning the
# reference to ONE kernel across it, is what makes the remaining comparison
# mean anything.
#
# This is also the shape sweep `notes/06` C-19 says this study got wrong before
# -- a threshold claim read at one shape. 2^22 elements is deliberately small:
# the claim being tested is about the SHAPE STRATEGY, and at this size every
# point is milliseconds so all eleven land inside the session.
%cd /kaggle/working/axisrepo
!DEV=CUDA python axis/harness.py --iters 20 --warmup 5 --skip-controls \
    --sweep-n 64,128,256,1024,2048,4096,8192 \
    --json /kaggle/working/axis_sweep.json 2>&1 | tail -70
```

```python
# ---- 9. bring the reports home as Kaggle outputs
#
# A MISSING REPORT IS A RESULT, NOT A NOTE. In version 18 cell 7 died of a
# one-character typo, this cell printed `axis_controls.json MISSING` and
# continued, and the fetched output was a clean-looking set of files with the
# controls silently absent. The ladder next to it was complete and correct, so
# nothing downstream could tell. So the absence of a file that a previous cell
# was supposed to write is raised here, loudly, with the cell that owed it.
import json
from sys import exit as _exit
from pathlib import Path
W = Path("/kaggle/working")

# cell -> the report it owes. Cell 9 is the last cell, so a missing file here
# cannot be produced by a later cell and is always a failure above.
OWED = {
    "axis_report.json":   "cell 6, the ladder",
    "axis_controls.json": "cell 8, the controls",
    "axis_sweep.json":    "cell 7, the width sweep (the falsifier)",
    "axis_smoke.json":    "cell 5, the smoke pass",
}
missing = []
for name, who in OWED.items():
    p = W / name
    size = p.stat().st_size if p.exists() else "MISSING"
    print(f"  {name:<20} {size:<10} owed by {who}")
    if not p.exists() or p.stat().st_size == 0:
        missing.append((name, who))

for name in ("axis_report.json", "axis_controls.json", "axis_sweep.json"):
    p = W / name
    if not p.exists():
        continue
    d = json.loads(p.read_text())
    print(f"\n---- {name} ----")
    if "sweep_scan_N" in d:
        print("  the falsifier: tg ns/element across mixin/op.py's SPLIT*2=512")
        print(f"  {'N':>7} {'R':>8} {'tg ms':>10} {'ref ms':>10} {'ratio':>8} "
              f"{'tg ns/el':>10} {'ref ns/el':>10} {'k':>5}")
        for row in d["sweep_scan_N"]:
            def num(v, w, p=4):
                return format(v, f".{p}f") if isinstance(v, (int, float)) else "-"
            ratio = row.get("ratio")
            print(f"  {row['N']:>7} {row['R']:>8} "
                  f"{num(row.get('tg_ms'), 10):>10} "
                  f"{num(row.get('ref_ms'), 10):>10} "
                  f"{(format(ratio, '.2f') + 'x' if ratio else '-'):>8} "
                  f"{num(row.get('ns_per_elem_tg'), 10):>10} "
                  f"{num(row.get('ns_per_elem_ref'), 10):>10} "
                  f"{str(row.get('kernels', '-')):>5}")
    if "build" in d:
        print(json.dumps(d.get("build", {}), indent=2)[:700])
        print("  floor  ", json.dumps(d.get("floor", {})))
        print("  device ", json.dumps(d.get("device", {})))
    for r in d.get("rungs", []):
        dev_ms = r.get("tg_device", {}).get("median_ms")
        rf = r.get("ref", {}).get("median_ms")
        print(f"  {r.get('key','?'):<9} tg_dev={dev_ms} ref={rf} "
              f"ratio={r.get('ratio_tg_over_ref')} {r.get('status')}")
if (W / "axis_run.log").exists():
    print("\n---- tail of axis_run.log ----")
    print((W / "axis_run.log").read_text()[-6000:])

if missing:
    print("\n" + "=" * 70)
    print("INCOMPLETE RUN -- these reports were never written:")
    for name, who in missing:
        print(f"  {name}  (owed by {who})")
    print("Any ratio below is real, but the run is not the run this notebook")
    print("describes. The note must not present a partial run as a complete one.")
    print("=" * 70)
    _exit(1)
print("\nall owed reports present.")
```
