```python

import base64, hashlib, json, os, subprocess, sys, time

OUT = "/kaggle/working"
os.makedirs(OUT, exist_ok=True)
EXPECT = ["6f376400341dd1015acd5f32e687e2fed7a85b4a26d4269b71680ec1f0c0d7d5", "752ffaea2028475df25e81eaa186ce8ae089e36eed53d96eeecd531663dabed9", "6c75c0e1e9cfc105e152f06c63c7695eff08781b9b8aa3e49ba098fe299ffbc3", "fe85581f80ed7fdba57398a6c04fe744a7fd59917d99b132ea0eb2dcfa457952", "d3d6356552db2be483430751924970d288cb1c128a56709aa074d7e18a908108", "7ec6927c711e0eec2b4de27d8eb03bb6c2f7e0d660edd8dfa969dba61434c539", "cecf6f98be10403115d4db675ad31d2061d79dcab01370e607467ecfe288ef6d", "85b5a0b5fb843e9791b3ef8303a88724483be51544ee18af795f200ffe152419", "c0d94bc0c822ccc490e5dbb185622aa9f37edd086f7786a56ba4c4b8c94e4e5d", "111b32e7a094d6456e1dbf08b81111bf4255487624bdb09da939c16f27173917", "d5b92e7eccf077deb7db8f70081ad9f354fb0c3dc28cff41d1121102815f7c5d", "f5ff7fe2e15f7d9b1313484187747da987c0d5fc165ec35b30daf3b10fc049bd", "c6ed8f06ee64d40a8a8ab7ebce3b3ceea1e3ac93f9ee50cfc1e63d3f090d878f", "d3a0931d642d67bd5e6fbe22e77a732da72edc3502ca8296f87f615a224965e8"]
PAYLOADS = [["minecu_tg/__init__.py", "IiIibWluZWN1X3RnIC0tIHRoZSBzYW1lIHZveGVsIEREQSwgd3JpdHRlbiBhcyBwbGFpbiBUaW55R3JhZC4KCkEgaGFuZC13cml0dGVuIENVREEgdm94ZWwgcmF5bWFyY2hlciBleGlzdHMgYXQgYG1pbmVjdV9yZWYvc3JjL2tlcm5lbHMuY3VgCigzLDI0NSBsaW5lcywgdW5tb2RpZmllZCkuIFRoaXMgcGFja2FnZSBpcyB0aGUgc2FtZSBjb21wdXRhdGlvbiB3cml0dGVuIHRoZSB3YXkKYSBwZXJzb24gd3JpdGVzIGl0IHdpdGggYXJyYXkgb3BlcmF0aW9ucyBhbmQgYSBsb29wLCBzbyB0aGUgc3R1ZHkgY2FuIG1lYXN1cmUKd2hlcmUgYSBjb2RlIGdlbmVyYXRvcidzIG91dHB1dCBsYW5kcyByZWxhdGl2ZSB0byBoYW5kLXdyaXR0ZW4gY29kZS4KCiAgICByZW5kZXIucHkgICAgICAgICB0aGUgRERBIHJlbmRlciwgYXMgdGVuc29yIG9wcyAgICAgICAgICA8LSB0aGUgc3ViamVjdAogICAgYmxvY2tzLnB5ICAgICAgICAgYmxvY2sgaWRzIGFuZCB0aGUgY29sb3VyIHRhYmxlLCBmcm9tIHRoZSByZWZlcmVuY2UKICAgIHdvcmxkLnB5ICAgICAgICAgIHRoZSB3b3JsZGdlbiBrZXJuZWxzLCBhcyB0ZW5zb3Igb3BzCiAgICBjcHVfcmVmZXJlbmNlLnB5ICBhbiBpbmRlcGVuZGVudCBmbG9hdDY0IENQVSB0cmFuc2NyaXB0aW9uCiAgICB2ZXJpZnkucHkgICAgICAgICB0aGUgY29ycmVjdG5lc3MgZ2F0ZQogICAgY29kZWdlbi5weSAgICAgICAga2VybmVsLWNvdW50IGNlbnN1cyBhbmQgQ1VEQSBzb3VyY2UgY2FwdHVyZQogICAgYmVuY2gucHkgICAgICAgICAgdGltaW5nIG9uIGEgcmVhbCBjYXJkCiAgICBrYWdnbGUvICAgICAgICAgICB0aGUgcHVzaCByZWNpcGUKClJlYWQgYHJlbmRlci5weWAncyBkb2NzdHJpbmcgYmVmb3JlIHF1b3RpbmcgYW55dGhpbmcgYWJvdXQgd2hhdCB0aGlzIGNvZGUgZG9lczsKdGhlIGNvbnN0cmFpbnRzIHRoYXQgc2hhcGVkIGl0IGFyZSBwYXJ0IG9mIHRoZSByZXN1bHQuCgpOT1RFIE9OIFRIRSBOQU1FUy4gYHJlbmRlcmAgaXMgdGhlICptb2R1bGUqLCBhbmQgYHJlbmRlci5yZW5kZXJgIGlzIHRoZQpmdW5jdGlvbi4gVGhpcyBwYWNrYWdlIGRlbGliZXJhdGVseSBkb2VzIG5vdCByZS1leHBvcnQgdGhlIGZ1bmN0aW9uIHVuZGVyIHRoZQpiYXJlIG5hbWUgYHJlbmRlcmAsIGJlY2F1c2UgdGhhdCB3b3VsZCBvdmVyd3JpdGUgdGhlIG1vZHVsZSBhdHRyaWJ1dGUgYW5kIG1ha2UKYG1pbmVjdV90Zy5yZW5kZXIuY2FtZXJhX2Jhc2lzYCBhIGNvbmZ1c2luZyBBdHRyaWJ1dGVFcnJvciBsYXRlci4gVXNlCgogICAgZnJvbSBtaW5lY3VfdGcucmVuZGVyIGltcG9ydCByZW5kZXIsIGNhbWVyYV9iYXNpcwoKb3IgdGhlIGByZW5kZXJfb2JzYCBhbGlhcyBiZWxvdy4KIiIiCgpmcm9tIC5yZW5kZXIgaW1wb3J0IGNhbWVyYV9iYXNpcwpmcm9tIC5yZW5kZXIgaW1wb3J0IHJlbmRlciBhcyByZW5kZXJfb2JzCmZyb20gLndvcmxkIGltcG9ydCBleHBhbmRfd29ybGQsIGdlbmVyYXRlX3dvcmxkLCBncm91bmRfb2YKCl9fYWxsX18gPSBbImNhbWVyYV9iYXNpcyIsICJyZW5kZXJfb2JzIiwgImdlbmVyYXRlX3dvcmxkIiwgImV4cGFuZF93b3JsZCIsICJncm91bmRfb2YiXQo=", "6f376400341dd1015acd5f32e687e2fed7a85b4a26d4269b71680ec1f0c0d7d5"], ["minecu_tg/blocks.py", "IiIiQmxvY2sgaWRzIGFuZCB0aGUgY29sb3VyIHRhYmxlcywgdHJhbnNjcmliZWQgZnJvbSB0aGUgcmVmZXJlbmNlLgoKRXZlcnl0aGluZyBoZXJlIGlzIGNvcGllZCBmcm9tIGBtaW5lY3VfcmVmL3NyYy9rZXJuZWxzLmN1YCwgd2hpY2ggaXMgdGhlCmhhbmQtd3JpdHRlbiBzaWRlIGFuZCBpcyBuZXZlciBtb2RpZmllZC4gTGluZSBudW1iZXJzIGluIHRoZSBjb21tZW50cyBhcmUgdGhlCnJlZmVyZW5jZSdzLgoKICAgIGtlcm5lbHMuY3U6MTEtMjQgICAgdGhlIDEzIGJsb2NrIGlkcywgQUlSID0gLTEKICAgIGtlcm5lbHMuY3U6NDc5LTQ5MyAgQkxPQ0tfQ09MT1JTX1U4CiAgICBrZXJuZWxzLmN1OjQ5NSAgICAgIFNLWV9DT0xPUl9VOAoiIiIKCmZyb20gdGlueWdyYWQgaW1wb3J0IFRlbnNvciwgZHR5cGVzCgpBSVIgPSAtMQpHUkFTUywgRElSVCwgU1RPTkUsIE9BS0xPRywgTEVBVkVTLCBTQU5ELCBXQVRFUiwgR0xBU1MsIEJSSUNLLCBDT0JCTEVTVE9ORSwgXAogICAgUExBTktTLCBTTk9XLCBCRURST0NLID0gcmFuZ2UoMTMpCgpCTE9DS19OQU1FUyA9ICgiR1JBU1MiLCAiRElSVCIsICJTVE9ORSIsICJPQUtMT0ciLCAiTEVBVkVTIiwgIlNBTkQiLCAiV0FURVIiLAogICAgICAgICAgICAgICAiR0xBU1MiLCAiQlJJQ0siLCAiQ09CQkxFU1RPTkUiLCAiUExBTktTIiwgIlNOT1ciLCAiQkVEUk9DSyIpCgojIGtlcm5lbHMuY3U6NDc5CkJMT0NLX0NPTE9SU19VOCA9IFsKICAgIFs3NiwgMTY2LCA1MV0sICAgICAgIyBHUkFTUwogICAgWzE0MCwgODksIDUxXSwgICAgICAjIERJUlQKICAgIFsxMjcsIDEyNywgMTI3XSwgICAgIyBTVE9ORQogICAgWzE0MCwgMTAyLCA2NF0sICAgICAjIE9BS0xPRwogICAgWzUxLCAxNDAsIDM4XSwgICAgICAjIExFQVZFUwogICAgWzIyOSwgMjE3LCAxNTNdLCAgICAjIFNBTkQKICAgIFs1MSwgMTAyLCAyMDRdLCAgICAgIyBXQVRFUgogICAgWzIxNywgMjI5LCAyNDJdLCAgICAjIEdMQVNTCiAgICBbMTc4LCA4OSwgNzZdLCAgICAgICMgQlJJQ0sKICAgIFsxMDIsIDEwMiwgMTAyXSwgICAgIyBDT0JCTEVTVE9ORQogICAgWzE5MSwgMTUzLCAxMDJdLCAgICAjIFBMQU5LUwogICAgWzI0MiwgMjQyLCAyNTBdLCAgICAjIFNOT1cKICAgIFszOCwgMzgsIDM4XSwgICAgICAgIyBCRURST0NLCl0KCiMga2VybmVscy5jdTo0OTUKU0tZX0NPTE9SX1U4ID0gWzEzNSwgMjA2LCAyMzVdCgoKZGVmIGNvbG91cl9sdXQoKToKICAgICIiIlRoZSBjb2xvdXIgdGFibGUgYXMgb25lIFsxNCwgM10gdWludDggdGFibGUuCgogICAgUm93IDAgaXMgR1JBU1MgLi4gcm93IDEyIGlzIEJFRFJPQ0ssIGV4YWN0bHkgYXMgYEJMT0NLX0NPTE9SU19VOGAgaXMKICAgIGluZGV4ZWQgaW4gdGhlIHJlZmVyZW5jZSAoYEJMT0NLX0NPTE9SU19VOFtoaXRfYmxvY2tdYCwgd2l0aCBoaXRfYmxvY2sgaW4KICAgIFswLCAxMl0pLiBSb3cgMTMgaXMgdGhlIHNreSwgd2hpY2ggdGhlIHJlZmVyZW5jZSByZWFjaGVzIGJ5IGEgKmJyYW5jaCoKICAgIHJhdGhlciB0aGFuIGFuIGluZGV4OiBgaGl0X2Jsb2NrYCBpcyBpbml0aWFsaXNlZCB0byBBSVIgPT0gLTEgYW5kCiAgICBgb3V0cHV0ID0gU0tZX0NPTE9SX1U4YCB3aGVuIHRoZSBEREEgZmluZHMgbm90aGluZy4gQXBwZW5kaW5nIHRoZSBza3kgYXMKICAgIHJvdyAxMyB0dXJucyB0aGF0IGJyYW5jaCBpbnRvIHRoZSBzYW1lIHRhYmxlIGxvb2t1cCB0aGUgaGl0IHBhdGggYWxyZWFkeQogICAgZG9lcywgc28gImRpZCBub3QgaGl0IiBhbmQgImhpdCIgYmVjb21lIG9uZSBnYXRoZXIuIFRoYXQgaXMgYSBjaGFuZ2Ugb2YKICAgICpzaGFwZSosIG5vdCBvZiBzY2hlZHVsZTogbm8gY29kZWdlbiBrbm9iLCBubyBmdXNpb24gaGludCwgbm8gdGlsZSBzaXplLgogICAgSXQgaXMgd2hhdCBhIHBlcnNvbiB3cml0aW5nIHRoZSB0ZW5zb3IgdmVyc2lvbiB3b3VsZCBkbyB0byBhdm9pZCBhbgogICAgYGlmYCBvbiBhIHRlbnNvci4KICAgICIiIgogICAgcm93cyA9IFtsaXN0KG1hcChpbnQsIGMpKSBmb3IgYyBpbiBCTE9DS19DT0xPUlNfVThdICsgW2xpc3QobWFwKGludCwgU0tZX0NPTE9SX1U4KSldCiAgICByZXR1cm4gVGVuc29yKHJvd3MsIGR0eXBlPWR0eXBlcy51aW50OCkK", "752ffaea2028475df25e81eaa186ce8ae089e36eed53d96eeecd531663dabed9"], ["minecu_tg/world.py", "IiIiVGhlIHZveGVsIHdvcmxkLCBhcyB0ZW5zb3Igb3BzLgoKVHJhbnNjcmliZWQgZnJvbSB0aHJlZSBoYW5kLXdyaXR0ZW4ga2VybmVscywgdW5tb2RpZmllZDoKCiAgICBnZW5lcmF0ZV93b3JsZF9rZXJuZWwgICBrZXJuZWxzLmN1OjE3OTMKICAgIHBsYWNlX3RyZWVfa2VybmVsICAgICAgIGtlcm5lbHMuY3U6MTgzMQogICAgcGxhY2VfaG91c2Vfa2VybmVsICAgICAga2VybmVscy5jdToxODY4CgpXcml0dGVuIHRoZSB3YXkgYSBwZXJzb24gd3JpdGVzIGl0IHdpdGggYXJyYXkgb3BlcmF0aW9uczogYnJvYWRjYXN0IHRoZSB0aHJlZQpjb29yZGluYXRlIGF4ZXMgYWdhaW5zdCBlYWNoIG90aGVyLCBidWlsZCB0aGUgYmxvY2sgaWQgd2l0aCBuZXN0ZWQgYHdoZXJlYCwKdGhlbiBzdGFtcCB0aGUgdHJlZSBhbmQgdGhlIGhvdXNlIGJ5IHNjYXR0ZXJpbmcgYSBoYW5kZnVsIG9mIGNvbnN0YW50IHZveGVscy4KClR3byBwcm9wZXJ0aWVzIG9mIHRoZSByZWZlcmVuY2UgYXJlIHdvcnRoIHN0YXRpbmcgYmVjYXVzZSB0aGV5IGFyZSBsb2FkLWJlYXJpbmcKZm9yIHRoZSBjb21wYXJpc29uIGFuZCBuZWl0aGVyIGlzIGEgY29kZWdlbiBjaG9pY2U6CgogICogKipUaGUgd29ybGQgaXMgaWRlbnRpY2FsIGZvciBldmVyeSBlbnYuKiogYGdlbmVyYXRlX3dvcmxkX2tlcm5lbGAgYWNjZXB0cyBhCiAgICBgc2VlZGAgYXJndW1lbnQgYW5kIG5ldmVyIHJlYWRzIGl0OyB0aGUgdHJlZSBhbmQgdGhlIGhvdXNlIGFyZSBwbGFjZWQgYXQKICAgIGZpeGVkIG9mZnNldHMuIFNvIGBnZW5lcmF0ZV93b3JsZGAgcmV0dXJucyBvbmUgW1ksIFgsIFpdIHZvbHVtZSBhbmQKICAgIGBleHBhbmRgIGdpdmVzIGV2ZXJ5IGVudiB0aGUgc2FtZSBvbmUuIFRoZSByZWZlcmVuY2UgbWF0ZXJpYWxpc2VzIEIKICAgIGlkZW50aWNhbCBjb3BpZXMgYW5kIHRoZSByZW5kZXIga2VybmVsIHJlYWRzIEIgc2VwYXJhdGUgYWRkcmVzc2VzOyB0aGlzCiAgICBtYXR0ZXJzIGZvciB0aGUgbWVtb3J5LXRyYWZmaWMgaGFsZiBvZiB0aGUgY29tcGFyaXNvbiBhbmQgaXMgY2FsbGVkIG91dAogICAgdGhlcmUgcmF0aGVyIHRoYW4gYmVpbmcgcGFwZXJlZCBvdmVyLgoKICAqICoqQXQgd29ybGQ9MTYgdGhlIGhvdXNlIGRvZXMgbm90IGZpdC4qKiBJdHMgY29ybmVyIGlzCiAgICBgKHNwYXduX3ggKyA2LCBncm91bmQsIHNwYXduX3ogLSAyKWAgPSBgKDE0LCBncm91bmQsIDYpYCBhbmQgaXQgaXMgNSB3aWRlLAogICAgc28gYGR4YCBpbiB7MiwgMywgNH0gbGFuZCBhdCB4IGluIHsxNiwgMTcsIDE4fSBhbmQgYXJlIGRyb3BwZWQgYnkgdGhlCiAgICBib3VuZHMgdGVzdC4gYGRkYV9tb2RlbC5weTo3MS0xMDNgIHNheXMgdGhlIHNhbWUgdGhpbmcgYWJvdXQgdGhlIHNhbWUKICAgIHdvcmxkLiBUaGUgaGFsZi1ob3VzZSBpcyB3aGF0IHRoZSByZWZlcmVuY2UgcmVuZGVycywgc28gdGhlIHRlbnNvciB2ZXJzaW9uCiAgICBidWlsZHMgYSBoYWxmLWhvdXNlIHRvbzsgdGhlIGJvdW5kcyB0ZXN0IGlzIHdyaXR0ZW4gb3V0IHJhdGhlciB0aGFuIGFzc3VtZWQuCiIiIgoKZnJvbSB0aW55Z3JhZCBpbXBvcnQgVGVuc29yLCBkdHlwZXMKCmZyb20gLmJsb2NrcyBpbXBvcnQgQUlSLCBCRURST0NLLCBESVJULCBHUkFTUywgT0FLTE9HLCBMRUFWRVMsIENPQkJMRVNUT05FLCBQTEFOS1MsIEJSSUNLCgoKZGVmIGdyb3VuZF9vZih3b3JsZDogaW50LCBncm91bmQ6IGludCB8IE5vbmUgPSBOb25lKSAtPiBpbnQ6CiAgICAiIiJUaGUgZ3JvdW5kIGhlaWdodCwgZGVyaXZlZCBmcm9tIHRoZSB3b3JsZCBzaXplIHdoZW4gbm90IGdpdmVuLgoKICAgIGBwaHlzaWNzX2tlcm5lbGAgZmxvb3JzIHRoZSBhZ2VudCBvbiBhIHBsYW5lIGF0IGB3b3JsZF95IC8gNGAgKmludGVnZXIqCiAgICBkaXZpc2lvbiAoa2VybmVscy5jdToxNTY2KSBhbmQgYGdlbmVyYXRlX3dvcmxkX2tlcm5lbGAgcHV0cyB0aGUgZ3Jhc3MgYXQKICAgIGBncm91bmRfaGVpZ2h0YCAoa2VybmVscy5jdToxODIzKS4gVGhlIHR3byBhZ3JlZSBvbmx5IHdoZW4KICAgIGBncm91bmRfaGVpZ2h0ID09IHdvcmxkX3kgLyA0YCwgYW5kIGRpc2FncmVlaW5nIGJ1cmllcyB0aGUgYWdlbnQgaW5zaWRlCiAgICBpdHMgb3duIHN1cmZhY2UsIHdoaWNoIGNvbGxhcHNlcyB0aGUgRERBIHRvIH4xLjMgc3RlcHMvcGl4ZWwuIERlcml2aW5nIGl0CiAgICBpcyB0aGUgY29uZmlndXJhdGlvbiB0aGUgYmFzZWxpbmUgd2FzIG1lYXN1cmVkIGluCiAgICAoYG1pbmVjdV9yZWYvYnVyaWVkX2FnZW50X2ZpeC5tZGApLgogICAgIiIiCiAgICByZXR1cm4gd29ybGQgLy8gNCBpZiBncm91bmQgaXMgTm9uZSBlbHNlIGdyb3VuZAoKCmRlZiBfYXhpcyhuOiBpbnQsIGRpbTogaW50KSAtPiBUZW5zb3I6CiAgICByZXR1cm4gVGVuc29yKGxpc3QocmFuZ2UobikpLCBkdHlwZT1kdHlwZXMuaW50MzIpLnJlc2hhcGUoCiAgICAgICAgbiBpZiBkaW0gPT0gMCBlbHNlIDEsIG4gaWYgZGltID09IDEgZWxzZSAxLCBuIGlmIGRpbSA9PSAyIGVsc2UgMSkKCgpkZWYgZ2VuZXJhdGVfd29ybGQod29ybGQ6IGludCwgZ3JvdW5kOiBpbnQgfCBOb25lID0gTm9uZSkgLT4gVGVuc29yOgogICAgIiIiT25lIFtZLCBYLCBaXSBpbnQ4IHZvbHVtZSwgZXhhY3RseSBhcyB0aGUgdGhyZWUgd29ybGRnZW4ga2VybmVscyBsZWF2ZSBpdC4KCiAgICBnZW5lcmF0ZV93b3JsZF9rZXJuZWwgaXMgYSB0aHJlZS13YXkgYnJhbmNoIG9uIHRoZSB5IGNvb3JkaW5hdGU6CgogICAgICAgIHkgPT0gMCAgICAgICAgICAgICAgICAgLT4gQkVEUk9DSwogICAgICAgIHkgPCAgZ3JvdW5kX2hlaWdodCAtIDEgLT4gRElSVAogICAgICAgIHkgPT0gZ3JvdW5kX2hlaWdodCAtIDEgLT4gR1JBU1MKICAgICAgICBvdGhlcndpc2UgICAgICAgICAgICAgIC0+IEFJUgoKICAgIHdoaWNoIGFzIHRlbnNvciBvcHMgaXMgdHdvIG5lc3RlZCBgd2hlcmVgcyBvdmVyIHRoZSB5IGF4aXMuIEVhY2ggYXhpcyBpcyBhCiAgICBbTiwgMSwgMV0gLyBbMSwgTiwgMV0gLyBbMSwgMSwgTl0gcm93IHNvIHRoZSB0aHJlZSBicm9hZGNhc3QgYWdhaW5zdCBlYWNoCiAgICBvdGhlciBpbnRvIFtZLCBYLCBaXS4KICAgICIiIgogICAgZyA9IGdyb3VuZF9vZih3b3JsZCwgZ3JvdW5kKQogICAgeSwgeCwgeiA9IF9heGlzKHdvcmxkLCAwKSwgX2F4aXMod29ybGQsIDEpLCBfYXhpcyh3b3JsZCwgMikKCiAgICBibGsgPSAoeSA9PSAwKS53aGVyZShCRURST0NLLCAoeSA8IGcgLSAxKS53aGVyZShESVJULCAoeSA9PSBnIC0gMSkud2hlcmUoR1JBU1MsIEFJUikpKQogICAgdm94ID0gKGJsayArIHggKiAwICsgeiAqIDApLmNhc3QoZHR5cGVzLmludDgpICAgIyBicm9hZGNhc3QgdG8gW1ksIFgsIFpdCgogICAgdm94ID0gX3BsYWNlX3RyZWUodm94LCB3b3JsZCwgZykKICAgIHZveCA9IF9wbGFjZV9ob3VzZSh2b3gsIHdvcmxkLCBnKQogICAgcmV0dXJuIHZveC5jb250aWd1b3VzKCkKCgpkZWYgX2F0KHZveDogVGVuc29yLCB5OiBpbnQsIHg6IGludCwgejogaW50LCBibG9jazogaW50KSAtPiBUZW5zb3I6CiAgICAiIiJPbmUgdm94ZWwgd3JpdGUsIGluIGJvdW5kcyBvciBub3QuCgogICAgVGhlIHJlZmVyZW5jZSdzIFNFVF9WT1hFTCBtYWNybyBib3VuZHMtY2hlY2tzIGJlZm9yZSB3cml0aW5nCiAgICAoa2VybmVscy5jdToxODk2KSwgc28gYSBob3VzZSB0aGF0IGRvZXMgbm90IGZpdCB3cml0ZXMgbm90aGluZyBvdXRzaWRlIHRoZQogICAgd29ybGQuIFNjYXR0ZXJpbmcgYSB3aG9sZSBbWSwgWCwgWl0gbWFzayBvZiBgYmxvY2tgIGludG8gYHZveGAgYW5kCiAgICBgd2hlcmVgLWluZyBpdCBpbiByZXByb2R1Y2VzIHRoYXQ6IGFuIG91dC1vZi1yYW5nZSBjb29yZGluYXRlIHNpbXBseSBuZXZlcgogICAgbWF0Y2hlcyBhbiBlbGVtZW50IG9mIHRoZSB2b2x1bWUuCiAgICAiIiIKICAgIHd5LCB3eCwgd3ogPSB2b3guc2hhcGUKICAgIHl5LCB4eCwgenogPSBfYXhpcyh3eSwgMCksIF9heGlzKHd4LCAxKSwgX2F4aXMod3osIDIpCiAgICBtID0gKCh5eSA9PSB5KSAmICh4eCA9PSB4KSAmICh6eiA9PSB6KSkKICAgIHJldHVybiBtLndoZXJlKGJsb2NrLCB2b3gpCgoKZGVmIF9wbGFjZV90cmVlKHZveDogVGVuc29yLCB3b3JsZDogaW50LCBncm91bmQ6IGludCkgLT4gVGVuc29yOgogICAgIiIicGxhY2VfdHJlZV9rZXJuZWwgKGtlcm5lbHMuY3U6MTgzMSk6IDMgT0FLTE9HIGF0IHkgPSBncm91bmQrZHkgLi4gKzIsIG9uZQogICAgTEVBVkVTIGF0IGdyb3VuZCszLCBhdCAoc3Bhd25feCwgc3Bhd25feiArIDIpLiBUaGUgcmVmZXJlbmNlJ3MgYGlmICh5IDwgd3kpYAogICAgaXMgdGhlIGJvdW5kcyB0ZXN0LCBzbyBhdCBhIHNtYWxsIHdvcmxkIHRoZSB0b3Agb2YgdGhlIHRyZWUgY2FuIGJlIGRyb3BwZWQuCiAgICAiIiIKICAgIHR4LCB0eiA9IHdvcmxkIC8vIDIsIHdvcmxkIC8vIDIgKyAyCiAgICBmb3IgZHkgaW4gcmFuZ2UoMyk6CiAgICAgICAgdm94ID0gX2F0KHZveCwgZ3JvdW5kICsgZHksIHR4LCB0eiwgT0FLTE9HKQogICAgdm94ID0gX2F0KHZveCwgZ3JvdW5kICsgMywgdHgsIHR6LCBMRUFWRVMpCiAgICByZXR1cm4gdm94CgoKZGVmIF9wbGFjZV9ob3VzZSh2b3g6IFRlbnNvciwgd29ybGQ6IGludCwgZ3JvdW5kOiBpbnQpIC0+IFRlbnNvcjoKICAgICIiInBsYWNlX2hvdXNlX2tlcm5lbCAoa2VybmVscy5jdToxODY4KTogNSB3aWRlIHggNCBkZWVwIHggNCB0YWxsIGF0CiAgICAoc3Bhd25feCArIDYsIHNwYXduX3ogLSAyKSwgYmFzZSB5ID0gZ3JvdW5kLiBGbG9vciBjb2JibGVzdG9uZSwgcGxhbmsgd2FsbHMKICAgIHdpdGggYSBkb29yIG9wZW5pbmcsIG9hayBjb3JuZXIgcGlsbGFycywgYnJpY2sgcm9vZi4iIiIKICAgIGh4LCBoeiwgVywgRCwgSCA9IHdvcmxkIC8vIDIgKyA2LCB3b3JsZCAvLyAyIC0gMiwgNSwgNCwgNAogICAgYmFzZSA9IGdyb3VuZAogICAgZm9yIGR4IGluIHJhbmdlKFcpOgogICAgICAgIGZvciBkeiBpbiByYW5nZShEKToKICAgICAgICAgICAgdm94ID0gX2F0KHZveCwgYmFzZSArIDAsIGh4ICsgZHgsIGh6ICsgZHosIENPQkJMRVNUT05FKQogICAgZm9yIGR5IGluIHJhbmdlKDEsIDQpOgogICAgICAgIGZvciBkeCBpbiByYW5nZShXKToKICAgICAgICAgICAgaWYgZHggPT0gMiBhbmQgZHkgPD0gMjoKICAgICAgICAgICAgICAgIGNvbnRpbnVlICAgICAgICAgICAgICAgICAgICAgICAgIyB0aGUgZG9vciBvcGVuaW5nCiAgICAgICAgICAgIHZveCA9IF9hdCh2b3gsIGJhc2UgKyBkeSwgaHggKyBkeCwgaHogKyAwLCBQTEFOS1MpCiAgICAgICAgICAgIHZveCA9IF9hdCh2b3gsIGJhc2UgKyBkeSwgaHggKyBkeCwgaHogKyBEIC0gMSwgUExBTktTKQogICAgICAgIGZvciBkeiBpbiByYW5nZSgxLCBEIC0gMSk6CiAgICAgICAgICAgIHZveCA9IF9hdCh2b3gsIGJhc2UgKyBkeSwgaHggKyAwLCBoeiArIGR6LCBQTEFOS1MpCiAgICAgICAgICAgIHZveCA9IF9hdCh2b3gsIGJhc2UgKyBkeSwgaHggKyBXIC0gMSwgaHogKyBkeiwgUExBTktTKQogICAgZm9yIGR5IGluIHJhbmdlKDEsIDQpOgogICAgICAgIGZvciBkeCwgZHogaW4gKCgwLCAwKSwgKFcgLSAxLCAwKSwgKDAsIEQgLSAxKSwgKFcgLSAxLCBEIC0gMSkpOgogICAgICAgICAgICB2b3ggPSBfYXQodm94LCBiYXNlICsgZHksIGh4ICsgZHgsIGh6ICsgZHosIE9BS0xPRykKICAgIGZvciBkeCBpbiByYW5nZShXKToKICAgICAgICBmb3IgZHogaW4gcmFuZ2UoRCk6CiAgICAgICAgICAgIHZveCA9IF9hdCh2b3gsIGJhc2UgKyBILCBoeCArIGR4LCBoeiArIGR6LCBCUklDSykKICAgIHJldHVybiB2b3gKCgpkZWYgZXhwYW5kX3dvcmxkKHZveDogVGVuc29yLCBiYXRjaDogaW50KSAtPiBUZW5zb3I6CiAgICAiIiJbWSwgWCwgWl0gLT4gW0IsIFksIFgsIFpdLCB0aGUgbGF5b3V0IGByZW5kZXJgIGdhdGhlcnMgZnJvbS4KCiAgICBgcmVwZWF0YCByYXRoZXIgdGhhbiBgZXhwYW5kYCArIGBjb250aWd1b3VzYCwgYmVjYXVzZSB0aGUgcmVuZGVyIGtlcm5lbAogICAgaW5kZXhlcyBgdm94ZWxzW2Iqd3h5eiArIC4uLl1gIGFuZCBhIG1hdGVyaWFsaXNlZCBjb3B5IGlzIHdoYXQgdGhlCiAgICByZWZlcmVuY2UgaGFzIGluIG1lbW9yeTsgYSBicm9hZGNhc3QgdmlldyB3b3VsZCBoYW5kIHRoZSBnYXRoZXIgYSBzdHJpZGUtMAogICAgYXhpcyBhbmQgcXVpZXRseSB0dXJuIGEgbWVtb3J5IHF1ZXN0aW9uIGludG8gYSBsYXlvdXQgdHJpY2suIFRoZSBjb3N0IG9mCiAgICB0aGF0IGNob2ljZSBpcyBCKndvcmxkXjMgYnl0ZXMgd3JpdHRlbiBvbmNlLCBhbmQgaXQgaXMgdGhlIHNhbWUgYnl0ZXMgdGhlCiAgICByZWZlcmVuY2Ugd3JpdGVzLgogICAgIiIiCiAgICByZXR1cm4gdm94LnJlc2hhcGUoMSwgKnZveC5zaGFwZSkucmVwZWF0KGJhdGNoLCAxLCAxLCAxKS5jb250aWd1b3VzKCkK", "6c75c0e1e9cfc105e152f06c63c7695eff08781b9b8aa3e49ba098fe299ffbc3"], ["minecu_tg/render.py", "IiIiVGhlIHJlbmRlciwgd3JpdHRlbiBhcyBwbGFpbiB0ZW5zb3IgY29kZS4KClRoaXMgaXMgdGhlIGZpbGUgdGhlIHdob2xlIHN0dWR5IHJlc3RzIG9uLCBzbyB0aGUgY29uc3RyYWludCB0aGF0IHByb2R1Y2VkIGl0CmlzIHdvcnRoIHdyaXRpbmcgZG93biBhdCB0aGUgdG9wOiAqKnRoaXMgaXMgdGhlICJwbGFpbiBDKysiIHNpZGUgb2YgdGhlCmNvbXBhcmlzb24uKiogSXQgaXMgd2hhdCBhIHBlcnNvbiB3cml0ZXMgd2hlbiB0aGV5IGFyZSBoYW5kZWQKYG5vdGVzLzE3LW1pbmVjdS1zcGVjLm1kYCBhbmQgdG9sZCB0byBpbXBsZW1lbnQKYHJlbmRlcl9rZXJuZWxfdWludDhfcHJlYmFzaXNgIHdpdGggYXJyYXkgb3BlcmF0aW9ucyBhbmQgYSBsb29wLiBJdCBpcyBub3QKd2hhdCBhIHBlcnNvbiB3cml0ZXMgYWZ0ZXIgcmVhZGluZyB0aGUgcmVmZXJlbmNlJ3MgU0FTUy4KClRoZXJlZm9yZSwgYW5kIHNwZWNpZmljYWxseSwgdGhpcyBmaWxlIGRvZXMgTk9UOgoKICAqIGNhbGwgYFRlbnNvci5jdXN0b21fa2VybmVsYCBvciBhbnkgb3RoZXIgZXNjYXBlIGhhdGNoIHRoYXQgZW1pdHMgQ1VEQQogICogc2V0IGEgdGlsZSBzaXplLCBgVENfU0VMRUNUYCwgYEFMTE9XX1RGMzJgLCBhIGJsb2NrIHNpemUsIG9yIGFueSBvdGhlcgogICAga25vYiB0aGUgY29kZSBnZW5lcmF0b3IgcmVhZHMKICAqIHVucm9sbCB0aGUgRERBIGxvb3AgYnkgaGFuZCwgZnVzZSBzdGFnZXMgYnkgaGFuZCwgb3Igc3RhZ2UgZGF0YSB0aHJvdWdoCiAgICBnbG9iYWwgbWVtb3J5IHRvIHNoYXBlIHRoZSBzY2hlZHVsZQogICogcGFzcyBhIGBDb250ZXh0KC4uLilgIHRvIHN0ZWVyIHRoZSBzY2hlZHVsZXIKICAqIHNwZWNpYWwtY2FzZSBgbWF4X3N0ZXBzYCAoZS5nLiBwZWVsIGEgMTYtd2lkZSBib2R5KSB0byBpbWl0YXRlIHRoZQogICAgcmVmZXJlbmNlJ3MgYCNwcmFnbWEgdW5yb2xsIDRgCgpUaGUgRERBIGxvb3AgaXMgYSBQeXRob24gYGZvcmAgb3ZlciBgbWF4X3N0ZXBzYCB3aXRoIG9uZSBib2R5LCBleGFjdGx5IGFzIG9uZQp3b3VsZCB3cml0ZSBhIHZlY3RvcmlzZWQgcmF5bWFyY2hlci4gV2hhdCBUaW55R3JhZCBkb2VzIHdpdGggdGhhdCB1bnJvbGxlZCBvcApjaGFpbiAtLSBmdXNlcyBpdCBpbnRvIG9uZSBrZXJuZWwsIHNwbGl0cyBpdCBpbnRvIGBtYXhfc3RlcHNgIGtlcm5lbHMsIG9yCmFueXRoaW5nIGluIGJldHdlZW4gLS0gaXMgdGhlIG1lYXN1cmVtZW50LCBub3QgYSBjaG9pY2UgbWFkZSBoZXJlLgoKVEhFIE9ORSBQTEFDRSBUSEUgU0hBUEUgSVMgQ0hPU0VOLCBBTkQgV0hZIElUIElTIE5PVCBBIFNDSEVEVUxFIENIT0lDRQotLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0KVGhlIHJlZmVyZW5jZSdzIGlubmVyIGxvb3AgaXMgZGF0YS1kZXBlbmRlbnQgcGVyIHJheTogYGlmIChibG9jayA+PSAwKSB7IC4uLgpnb3RvIGRvbmU7IH1gLiBBIHRlbnNvciBwcm9ncmFtIGhhcyBubyBgZ290b2AsIGFuZCB0aGUgaG9uZXN0IHdheSB0byBleHByZXNzCiJ0aGlzIHJheSBpcyBmaW5pc2hlZCIgaW4gYXJyYXkgY29kZSBpcyB0byBrZWVwIG1hcmNoaW5nIGFuZCBzdG9wICphY2NlcHRpbmcKcmVzdWx0cyogLS0gY2FycnkgYW4gYGFjdGl2ZWAgbWFzaywgYW5kIG1ha2UgZXZlcnkgdXBkYXRlIGEgYHdoZXJlKGFjdGl2ZSAmIGMsCm5ldywgb2xkKWAuIFRoYXQgaXMgdGhlIHN0YW5kYXJkIHZlY3RvcmlzZWQtcmF5bWFyY2hlciBpZGlvbSBhbmQgaXQgaXMgd2hhdAp0aGlzIGZpbGUgZG9lcy4gVGhlIGFsdGVybmF0aXZlLCBhIGB3aGlsZWAvY2FycnktdW50aWwtY29udmVyZ2VkIGZvcm0sIGlzCmV4cHJlc3NpYmxlIGluIFRpbnlHcmFkIG9ubHkgdGhyb3VnaCBsb3ctbGV2ZWwgbG9vcCBjb25zdHJ1Y3RzIHRoYXQgYXJlCmhhbmQtc3RhZ2luZywgc28gaXQgaXMgbm90IHVzZWQuCgpUaGUgY29uc2VxdWVuY2UgaXMgaG9uZXN0IGFuZCBpcyAqdGhlIGZpbmRpbmcqLCBub3QgYSBkZWZlY3Qgb2YgdGhlIHdyaXRpbmc6CmV2ZXJ5IHJheSBydW5zIHRoZSBmdWxsIGBtYXhfc3RlcHNgIGl0ZXJhdGlvbnMgd2hldGhlciBpdCBoaXRzIG9uIHN0ZXAgMSBvcgpzdGVwIDMyLCBzbyB0aGUgd29yayBpcyBgbWF4X3N0ZXBzYCBwZXIgcGl4ZWwgcmF0aGVyIHRoYW4gdGhlIG1lYXN1cmVkIG1lYW4gb2YKfjI2LiBUaGUgaGFuZC13cml0dGVuIGtlcm5lbCdzIGAjcHJhZ21hIHVucm9sbCA0YCAoMTYgc3RhdGljIHN0ZXBzLCAzOCBTQVNTCmluc3RydWN0aW9ucyBhcGFydCkgZXhpc3RzIHRvIGtlZXAgMTYgKmxvYWQgY2hhaW5zKiBpbiBmbGlnaHQsIHdoaWNoIGlzIGEKbWVtb3J5LWxldmVsLXBhcmFsbGVsaXNtIGFyZ3VtZW50IGFib3V0IG9uZSByYXkncyBkZXBlbmRlbnQgY2hhaW47IGEgbWFza2VkCnRlbnNvciBsb29wIGhhcyBubyBkZXBlbmRlbnQgY2hhaW4gdG8gb3ZlcmxhcCBiZWNhdXNlIGFsbCByYXlzIGFyZSBhbHJlYWR5IGluCmZsaWdodCB0b2dldGhlci4gVGhlc2UgYXJlIGRpZmZlcmVudCBtYWNoaW5lcyBkb2luZyB0aGUgc2FtZSBhcml0aG1ldGljLCBhbmQKdGhlIG5vdGVzIGZpbGUgbWVhc3VyZXMgYm90aCBhbmQgZG9lcyBub3QgcHJldGVuZCB0aGV5IGFyZSB0aGUgc2FtZSB0aGluZy4KCk5PIEVBUkxZIEVYSVQgUEVSIFJBWSwgQU5EIFdIWSBUSEFUIElTIFRIRSBQT0lOVAotLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0KYGZvciBpIGluIHJhbmdlKG1heF9zdGVwcylgIC0tIHRoZSBsb29wIGJvdW5kIGlzIHRoZSByZWZlcmVuY2UncyBgbWF4X3N0ZXBzYCwKaXRzIHBlci1yYXkgZWFybHkgZXhpdCBiZWNvbWVzIGFuIGBhY3RpdmVgIG1hc2ssIGFuZCB0aGUgd2hvbGUgYm9keSBpcyBvbmUKdW5yb2xsZWQgY2hhaW4uIElmIFRpbnlHcmFkIGZ1c2VzIGFjcm9zcyBpdGVyYXRpb25zIHdlIGdldCBvbmUga2VybmVsIHdpdGgKMzIgaW5saW5lZCBzdGVwczsgaWYgaXQgZG9lcyBub3QsIHdlIGdldCBvbmUga2VybmVsIHBlciBzdGVwIHdpdGggdGhlIHJheQpwb3NpdGlvbiBtYXRlcmlhbGlzZWQgaW4gYmV0d2Vlbi4gQm90aCBhcmUgb3V0Y29tZXMuIFRoZSBjb2RlIGRvZXMgbm90IGNob29zZQpiZXR3ZWVuIHRoZW0uCiIiIgoKZnJvbSB0aW55Z3JhZCBpbXBvcnQgVGVuc29yLCBkdHlwZXMKCmZyb20gLmJsb2NrcyBpbXBvcnQgY29sb3VyX2x1dCwgQUlSCgojIFRoZSByZWZlcmVuY2UncyBgMWUzMGZgIGd1YXJkIGZvciBhIHplcm8gcmF5IGNvbXBvbmVudCAoa2VybmVscy5jdTo2NDYtNjQ4KS4KX0JJRyA9IDFlMzAKCgpkZWYgY2FtZXJhX2Jhc2lzKGNhbWVyYXM6IFRlbnNvciwgZm92X2RlZ3JlZXM6IGZsb2F0LCB3aWR0aDogaW50LCBoZWlnaHQ6IGludCkgLT4gVGVuc29yOgogICAgIiIicHJlY29tcHV0ZV9jYW1lcmFfYmFzaXNfa2VybmVsIChrZXJuZWxzLmN1OjUxMSk6IFtCLCA1XSAtPiBbQiwgMTRdLgoKICAgIGNhbWVyYXMgaXMgKHgsIHksIHosIHlhdywgcGl0Y2gpLiBUaGUgb3V0cHV0IGxheW91dCBpcyB0aGUgcmVmZXJlbmNlJ3M6CgogICAgICAgIFswLi4yXSAgZXllLCB3aXRoIHkgcmFpc2VkIGJ5IHRoZSBleWUgaGVpZ2h0IDEuNgogICAgICAgIFszLi41XSAgZm9yd2FyZCA9IChzaW5feWF3KmNvc19waXRjaCwgc2luX3BpdGNoLCBjb3NfeWF3KmNvc19waXRjaCkKICAgICAgICBbNi4uN10gIHJpZ2h0LngsIHJpZ2h0Lnogb25seSAtLSByaWdodC55IGlzIGlkZW50aWNhbGx5IDAKICAgICAgICBbOC4uMTBdIHVwCiAgICAgICAgWzExXSAgICBzY2FsZV91ID0gaGFsZl9mb3YgKiBhc3BlY3QKICAgICAgICBbMTJdICAgIHNjYWxlX3YgPSBoYWxmX2ZvdgoKICAgIFRyYW5zY3JpYmVkIHN0YXRlbWVudCBmb3Igc3RhdGVtZW50LCBpbmNsdWRpbmcgdGhlIGRyb3BwZWQgcmlnaHQueSAoYSAzCiAgICB2ZWN0b3IgaW4gMiBzbG90cywgd2hpY2ggYSBjb21waWxlciB3b3VsZCBoYXZlIHRvIGJlIHRvbGQpIGFuZCB0aGUKICAgIGBmb3YgKiAwLjAxNzQ1MzI5ZmAgZGVncmVlLXRvLXJhZGlhbiBjb25zdGFudC4gYC0tdXNlX2Zhc3RfbWF0aGAgdHVybnMKICAgIHNpbmYvY29zZi90YW5mIGludG8gYXBwcm94aW1hdGUgaW50cmluc2ljcyBpbiB0aGUgcmVmZXJlbmNlOyBoZXJlIHRoZXkgYXJlCiAgICBUaW55R3JhZCdzIHNpbi9jb3MvdGFuLCB3aGljaCB0aGUgQ1VEQSBiYWNrZW5kIGxvd2VycyB0byB0aGUgc2FtZSBNVUZVCiAgICBzZXF1ZW5jZSB1cCB0byBhIGxhc3QtdWxwIGRpZmZlcmVuY2UuIFRoZSBEREEgY29tcGFyaXNvbiBpcyBhIGZsb2F0CiAgICBjb21wYXJlLCBzbyBhIGxhc3QtdWxwIGRpZmZlcmVuY2UgY2FuIG1vdmUgYSByYXkgb25lIHZveGVsIGF0IGFuIGV4YWN0IHRpZTsKICAgIHRoYXQgaXMgd2h5IGNvcnJlY3RuZXNzIGlzIHBpbm5lZCB0byBhIENQVSByZWZlcmVuY2UgYW5kIHRvIHRoZSBEREEgc3RlcAogICAgY291bnQgcmF0aGVyIHRoYW4gdG8gYSBiaXQtZXhhY3QgaW1hZ2UgYWdhaW5zdCB0aGUgR1BVLgogICAgIiIiCiAgICBmMzIgPSBkdHlwZXMuZmxvYXQzMgogICAgY2FtX3gsIGNhbV95LCBjYW1feiA9IGNhbWVyYXNbOiwgMF0sIGNhbWVyYXNbOiwgMV0sIGNhbWVyYXNbOiwgMl0KICAgIHlhdywgcGl0Y2ggPSBjYW1lcmFzWzosIDNdLCBjYW1lcmFzWzosIDRdCiAgICB6ZXJvID0geWF3ICogMCAgICAgICAgICAgICAgICAgICAgICAgIyBhIFtCXSB6ZXJvLCBmb3IgYnJvYWRjYXN0aW5nIGEgc2NhbGFyCgogICAgc2luX3lhdywgY29zX3lhdyA9IHlhdy5zaW4oKSwgeWF3LmNvcygpCiAgICBzaW5fcGl0Y2gsIGNvc19waXRjaCA9IHBpdGNoLnNpbigpLCBwaXRjaC5jb3MoKQoKICAgICMgdGFuZiBpbiBmbG9hdDMyIG9uIHRoZSBkZXZpY2UsIGFzIHRoZSByZWZlcmVuY2UgZG9lcyAtLSBub3QgbWF0aC50YW4gaW4KICAgICMgZG91YmxlLCB3aGljaCBpcyBhIGRpZmZlcmVudCBsYXN0LXVscCBiZWhhdmlvdXIgZnJvbSB0aGUgb25lIGJlaW5nCiAgICAjIGNvbXBhcmVkIGFnYWluc3QuCiAgICBoYWxmX2ZvdiA9IFRlbnNvcihbZm92X2RlZ3JlZXMgKiAwLjAxNzQ1MzI5ICogMC41XSwgZHR5cGU9ZjMyKS50YW4oKVswXSArIHplcm8KICAgIGFzcGVjdCA9IGZsb2F0KHdpZHRoKSAvIGZsb2F0KGhlaWdodCkKCiAgICBleWUgPSBjYW1feC5zdGFjayhjYW1feSArIDEuNiwgY2FtX3osIGRpbT0xKS5jYXN0KGYzMikKICAgIGZvcndhcmQgPSAoc2luX3lhdyAqIGNvc19waXRjaCkuc3RhY2soc2luX3BpdGNoLCBjb3NfeWF3ICogY29zX3BpdGNoLCBkaW09MSkKICAgIHJpZ2h0ID0gY29zX3lhdy5zdGFjaygtc2luX3lhdywgZGltPTEpICAgICAgICAgICAgICAgICAgICAjIHJpZ2h0X3kgPT0gMAogICAgdXAgPSAoLXNpbl95YXcgKiBzaW5fcGl0Y2gpLnN0YWNrKGNvc19waXRjaCwgLWNvc195YXcgKiBzaW5fcGl0Y2gsIGRpbT0xKQogICAgc2NhbGVzID0gKGhhbGZfZm92ICogYXNwZWN0KS5zdGFjayhoYWxmX2ZvdiwgZGltPTEpCiAgICByZXR1cm4gZXllLmNhdChmb3J3YXJkLCByaWdodCwgdXAsIHNjYWxlcywgZGltPTEpLmNvbnRpZ3VvdXMoKQoKCmRlZiByZW5kZXIodm94ZWxzOiBUZW5zb3IsIGJhc2lzOiBUZW5zb3IsIHdpZHRoOiBpbnQsIGhlaWdodDogaW50LAogICAgICAgICAgIG1heF9zdGVwczogaW50LCBjb3VudF9zdGVwczogYm9vbCA9IEZhbHNlKToKICAgICIiIlRoZSB1aW50OCBvYnNlcnZhdGlvbiwgW0IsIEgsIFcsIDNdLgoKICAgIHZveGVsczogW0IsIFdZLCBXWCwgV1pdIGludDguIGJhc2lzOiBbQiwgMTRdIGZsb2F0MzIsIGZyb20gYGNhbWVyYV9iYXNpc2AuCgogICAgVGhlIGJvZHkgaXMgdGhlIHJlZmVyZW5jZSdzIGByZW5kZXJfa2VybmVsX3VpbnQ4X3ByZWJhc2lzYAogICAgKGtlcm5lbHMuY3U6NTY4KSB3cml0dGVuIG92ZXIgYXJyYXlzLCBpbiB0aGUgb3JkZXIgdGhlIHJlZmVyZW5jZSBkb2VzIGl0OgogICAgYnVpbGQgdGhlIHJheSwgbm9ybWFsaXNlIGl0LCBzZXQgdXAgdGhlIGdyaWQgdHJhdmVyc2FsLCBsb29wLCB0aGVuIGxvb2sgdGhlCiAgICBibG9jayB1cCBpbiB0aGUgY29sb3VyIHRhYmxlLgoKICAgIGBjb3VudF9zdGVwc2AgcmV0dXJucyBgKG9icywgc3RlcHMpYCB3aGVyZSBgc3RlcHNgIGlzIFtCLCAxLCBILCBXXSBpbnQzMiwgdGhlCiAgICBudW1iZXIgb2YgRERBIGJvZGllcyBhIHJheSBleGVjdXRlZC4gSXQgaXMgKippbnN0cnVtZW50YXRpb24qKiwgbm90IHBhcnQgb2YKICAgIHRoZSBjb21wdXRhdGlvbjogaXQgaXMgdGhlIHNhbWUgbG9vcCB3aXRoIGEgY291bnRlciwgaXQgY2Fubm90IGNoYW5nZSBgb2JzYAogICAgKGB2ZXJpZnkucHlgIGFzc2VydHMgdGhlIHR3byBhcmUgYml0LWlkZW50aWNhbCksIGFuZCBldmVyeSB0aW1pbmcgaW4gdGhpcwogICAgc3R1ZHkgaXMgdGFrZW4gd2l0aCBpdCBvZmYuIEl0IGV4aXN0cyBiZWNhdXNlIHRoZSBzdGVwIGNvdW50IGlzIGEgZGlzY3JldGUsCiAgICB0aWUtc2Vuc2l0aXZlIHF1YW50aXR5OiB0d28gaW1wbGVtZW50YXRpb25zIHRoYXQgdGFrZSB0aGUgc2FtZSB0cmF2ZXJzYWwKICAgIGRlY2lzaW9ucyBoYXZlIGlkZW50aWNhbCBzdGVwIGNvdW50cywgYW5kIHR3byB0aGF0IGRvIG5vdCwgZG8gbm90LiBUaGUKICAgIGltYWdlIGlzIGEgbG9zc3kgcmVhZG91dCBvZiB0aGF0IC0tIG1hbnkgc3RlcCBjb3VudHMgY29sbGFwc2UgdG8gdGhlIHNhbWUKICAgIGNvbG91ciAtLSBzbyB0aGUgc3RlcCBjb3VudCBpcyB0aGUgc2hhcnBlciBvZiB0aGUgdHdvIGNoZWNrcy4KICAgICIiIgogICAgQiA9IHZveGVscy5zaGFwZVswXQogICAgd3ksIHd4LCB3eiA9IHZveGVscy5zaGFwZVsxXSwgdm94ZWxzLnNoYXBlWzJdLCB2b3hlbHMuc2hhcGVbM10KICAgIGYzMiwgaTMyID0gZHR5cGVzLmZsb2F0MzIsIGR0eXBlcy5pbnQzMgoKICAgICMgLS0tLSB0aGUgZXllIGFuZCB0aGUgYmFzaXMsIGxpZnRlZCB0byBbQiwgMSwgMSwgMV0gc28gdGhleSBicm9hZGNhc3QKICAgICMgb3ZlciB0aGUgcGl4ZWwgZ3JpZCB0aGUgd2F5IG9uZSBzY2FsYXIgZG9lcyBwZXIgdGhyZWFkIGluIHRoZSByZWZlcmVuY2UuCiAgICAjIFRoZSBzbGljZSBpcyBgaTppKzFgIGFuZCBub3QgYGlgOiBgYmFzaXNbOiwgaV1gIGRyb3BzIHRvIHJhbmsgMSwgYW5kCiAgICAjIGBbTm9uZSwgTm9uZV1gIG9uIGEgcmFuay0xIHRlbnNvciBnaXZlcyAoMSwgMSwgQikgLS0gdGhlIGJhdGNoIGxhbmRzIGluCiAgICAjIHRoZSB3cm9uZyBheGlzLCB3aGljaCBpcyBpbnZpc2libGUgYXQgQj0xIGFuZCB3cm9uZyBhdCBldmVyeSBvdGhlciBCLgogICAgZGVmIGNvbXAoaSk6CiAgICAgICAgcmV0dXJuIGJhc2lzWzosIGk6aSArIDEsIE5vbmUsIE5vbmVdCgogICAgZXllX3gsIGV5ZV95LCBleWVfeiA9IGNvbXAoMCksIGNvbXAoMSksIGNvbXAoMikKICAgIGZ3ZF94LCBmd2RfeSwgZndkX3ogPSBjb21wKDMpLCBjb21wKDQpLCBjb21wKDUpCiAgICByaWdodF94LCByaWdodF96ID0gY29tcCg2KSwgY29tcCg3KQogICAgdXBfeCwgdXBfeSwgdXBfeiA9IGNvbXAoOCksIGNvbXAoOSksIGNvbXAoMTApCiAgICBzY2FsZV91LCBzY2FsZV92ID0gY29tcCgxMSksIGNvbXAoMTIpCgogICAgIyAtLS0tIHRoZSBwaXhlbCBncmlkLiBweCBydW5zIGFsb25nIHRoZSBsYXN0IGF4aXMsIHB5IGFsb25nIHRoZSB0aGlyZCwgc28KICAgICMgdGhlIG91dHB1dCBpcyBhbHJlYWR5IFtCLCAxLCBILCBXXSBoZXJlIGFuZCBuZWVkcyBubyB0cmFuc3Bvc2UgbGF0ZXIuCiAgICBweCA9IFRlbnNvcihsaXN0KHJhbmdlKHdpZHRoKSksIGR0eXBlPWYzMikucmVzaGFwZSgxLCAxLCAxLCB3aWR0aCkKICAgIHB5ID0gVGVuc29yKGxpc3QocmFuZ2UoaGVpZ2h0KSksIGR0eXBlPWYzMikucmVzaGFwZSgxLCAxLCBoZWlnaHQsIDEpCgogICAgIyBrZXJuZWxzLmN1OjYxOS02MjAKICAgIHUgPSAoKHB4ICsgMC41KSAqICgyLjAgLyBmbG9hdCh3aWR0aCkpIC0gMS4wKSAqIHNjYWxlX3UKICAgIHYgPSAoMS4wIC0gKHB5ICsgMC41KSAqICgyLjAgLyBmbG9hdChoZWlnaHQpKSkgKiBzY2FsZV92CgogICAgIyBrZXJuZWxzLmN1OjYyMi02MjQuIHJpZ2h0X3kgaXMgMCwgc28gcmF5X3kgaGFzIG5vIHUgdGVybSAtLSB0aGUgc2FtZQogICAgIyBmYWN0IHRoZSByZWZlcmVuY2UgaGFyZC1jb2RlcyBpbiBpdHMgYmFzaXMgbGF5b3V0LgogICAgcmF5X3ggPSBmd2RfeCArIHUgKiByaWdodF94ICsgdiAqIHVwX3gKICAgIHJheV95ID0gZndkX3kgKyB2ICogdXBfeQogICAgcmF5X3ogPSBmd2RfeiArIHUgKiByaWdodF96ICsgdiAqIHVwX3oKCiAgICAjIGtlcm5lbHMuY3U6NjI2LTYyOSwgcnNxcnRmLiB0aW55Z3JhZCdzIGByc3FydGAgbG93ZXJzIHRvIHRoZSBzYW1lCiAgICAjIE1VRlUuUlNRIG9uIHNtXzc1LgogICAgcmF5X2xlbiA9IChyYXlfeCAqIHJheV94ICsgcmF5X3kgKiByYXlfeSArIHJheV96ICogcmF5X3opLnJzcXJ0KCkKICAgIHJheV94LCByYXlfeSwgcmF5X3ogPSByYXlfeCAqIHJheV9sZW4sIHJheV95ICogcmF5X2xlbiwgcmF5X3ogKiByYXlfbGVuCgogICAgIyAtLS0tIGdyaWQgdHJhdmVyc2FsIHNldHVwLCBrZXJuZWxzLmN1OjYzOC02NTkKICAgIHZveGVsX3ggPSBleWVfeC5mbG9vcigpCiAgICB2b3hlbF95ID0gZXllX3kuZmxvb3IoKQogICAgdm94ZWxfeiA9IGV5ZV96LmZsb29yKCkKCiAgICBzdGVwX3ggPSAocmF5X3ggPj0gMCkud2hlcmUoMS4wLCAtMS4wKQogICAgc3RlcF95ID0gKHJheV95ID49IDApLndoZXJlKDEuMCwgLTEuMCkKICAgIHN0ZXBfeiA9IChyYXlfeiA+PSAwKS53aGVyZSgxLjAsIC0xLjApCgogICAgIyAocmF5ICE9IDApID8gMS9yYXkgOiAxZTMwLiBUaGUgYHdoZXJlYCBldmFsdWF0ZXMgYm90aCBicmFuY2hlcyBhbmQgcGlja3MKICAgICMgb25lLCB3aGljaCBpcyB0aGUgc2FtZSB0cmFkZSB0aGUgcmVmZXJlbmNlJ3MgdGVybmFyeSBtYWtlczsgdGhlIGRpc2NhcmRlZAogICAgIyBicmFuY2ggaXMgYW4gaW5maW5pdHksIG5ldmVyIGEgTmFOLCBzbyBub3RoaW5nIGVzY2FwZXMuCiAgICBpbnZfeCA9IChyYXlfeCAhPSAwKS53aGVyZShyYXlfeC5yZWNpcHJvY2FsKCksIF9CSUcpCiAgICBpbnZfeSA9IChyYXlfeSAhPSAwKS53aGVyZShyYXlfeS5yZWNpcHJvY2FsKCksIF9CSUcpCiAgICBpbnZfeiA9IChyYXlfeiAhPSAwKS53aGVyZShyYXlfei5yZWNpcHJvY2FsKCksIF9CSUcpCgogICAgdF9kZWx0YV94LCB0X2RlbHRhX3ksIHRfZGVsdGFfeiA9IGludl94LmFicygpLCBpbnZfeS5hYnMoKSwgaW52X3ouYWJzKCkKCiAgICB0X21heF94ID0gKHJheV94ID49IDApLndoZXJlKCh2b3hlbF94ICsgMS4wIC0gZXllX3gpICogaW52X3gsICh2b3hlbF94IC0gZXllX3gpICogaW52X3gpCiAgICB0X21heF95ID0gKHJheV95ID49IDApLndoZXJlKCh2b3hlbF95ICsgMS4wIC0gZXllX3kpICogaW52X3ksICh2b3hlbF95IC0gZXllX3kpICogaW52X3kpCiAgICB0X21heF96ID0gKHJheV96ID49IDApLndoZXJlKCh2b3hlbF96ICsgMS4wIC0gZXllX3opICogaW52X3osICh2b3hlbF96IC0gZXllX3opICogaW52X3opCgogICAgIyAtLS0tIHRoZSBEREEsIGtlcm5lbHMuY3U6NjY0LTY3OQogICAgIwogICAgIyBgaGl0YCBzdGFydHMgYXQgQUlSIGFuZCBgZm91bmRgIHN0YXJ0cyBmYWxzZS4gRWFjaCBpdGVyYXRpb24gcmVhZHMgdGhlCiAgICAjIHZveGVsIHVuZGVyIHRoZSBjdXJzb3IsIHJlY29yZHMgdGhlIGhpdCBmb3IgYSByYXkgdGhhdCBpcyBzdGlsbCBsb29raW5nCiAgICAjIGFuZCBzdGlsbCBpbnNpZGUgdGhlIHdvcmxkLCB0aGVuIHN0ZXBzLiBBIHJheSB0aGF0IGhhcyBoaXQgc3RvcHMgY2hhbmdpbmcKICAgICMgYGhpdGA7IGl0IGtlZXBzIHN0ZXBwaW5nLCBoYXJtbGVzc2x5LCBiZWNhdXNlIGV2ZXJ5IHVwZGF0ZSBpcyBtYXNrZWQuCiAgICAjCiAgICAjIFRocmVlIHRoaW5ncyBhYm91dCB0aGlzIGxvb3AgYXJlIGRlbGliZXJhdGUgYW5kIG5vbmUgb2YgdGhlbSBpcyBhCiAgICAjIHNjaGVkdWxpbmcgZGVjaXNpb246CiAgICAjCiAgICAjICAqIGBmb3VuZGAgYW5kIGBpbnNpZGVgIGFyZSBzZXBhcmF0ZSBtYXNrcy4gVGhlIHJlZmVyZW5jZSdzIGBnb3RvIGRvbmVgCiAgICAjICAgIGZpcmVzIG9uIGEgaGl0OyBpdHMgYm91bmRzIHRlc3QgZ2F0ZXMgb25seSB0aGUgKmxvYWQqLiBBIHJheSB0aGF0CiAgICAjICAgIGxlYXZlcyB0aGUgd29ybGQgaGFzIG5vdCBoaXQgYW55dGhpbmcsIGFuZCBjb25mbGF0aW5nIHRoZSB0d28gd291bGQKICAgICMgICAgY29ycnVwdCB0aGUgc3RlcCBjb3VudCBmb3IgZXhhY3RseSB0aGUgcmF5cyB0aGF0IGRvbWluYXRlIHRoaXMKICAgICMgICAgd29ya2xvYWQgLS0gbW9zdCBvZiB0aGVtIHJ1biBvZmYgdGhlIGVuZCBvZiB0aGUgd29ybGQsIG5vdCBpbnRvIGEKICAgICMgICAgYmxvY2suCiAgICAjCiAgICAjICAqIGBpbnNpZGVgIGlzIHJlLWV2YWx1YXRlZCBldmVyeSBpdGVyYXRpb24sIGZyb20gdGhlIGxpdmUgY3Vyc29yLiBUaGUKICAgICMgICAgcmVmZXJlbmNlJ3MgYCh1bnNpZ25lZCl2b3hlbF94IDwgKHVuc2lnbmVkKXd4YCBpcyBpbnNpZGUgaXRzIGxvb3AgZm9yCiAgICAjICAgIHRoZSBzYW1lIHJlYXNvbjogdGhlIGN1cnNvciBtb3Zlcy4KICAgICMKICAgICMgICogdGhlIGdhdGhlciBpbmRleCBpcyBjbGFtcGVkLiBUaGF0IGlzIG1lbW9yeSBzYWZldHksIGJlY2F1c2UgYSB0ZW5zb3IKICAgICMgICAgZ2F0aGVyIGhhcyBubyBib3VuZHMgY2hlY2sgYW5kIHRoZSByZWZlcmVuY2UncyBib3VuZHMgdGVzdCBnYXRlcyBhCiAgICAjICAgIHBvaW50ZXIgaXQgc2ltcGx5IG5ldmVyIGZvcm1zLiBgaW5zaWRlYCBpcyB3aGF0IGRlY2lkZXMgd2hldGhlciB0aGUKICAgICMgICAgdmFsdWUgcmVhZCBjb3VudHMuCiAgICAjCiAgICAjICAqIHRoZSBnYXRoZXIgaXMgd3JpdHRlbiBhcyBhIDQtaW5kZXggYHZveGVsc1tiLCB5LCB4LCB6XWAsIHdoaWNoIGlzIHRoZQogICAgIyAgICBvYnZpb3VzIHdheSB0byBzYXkgaXQgYW5kIHdoaWNoIHRpbnlncmFkIGxvd2VycyB0byBmb3VyIHByZWRpY2F0ZWQKICAgICMgICAgMS1ieXRlIGxvYWRzIG9mIHRoZSAqc2FtZSogYWRkcmVzcyAtLSBvbmUgcGVyIGBib29sYCBjb25zdW1lciBvZiB0aGUKICAgICMgICAgc3RlcCwgYmVjYXVzZSB0aGUgY29tcGlsZXIgY2Fubm90IHByb3ZlIHRoZSBmb3VyIHByZWRpY2F0ZXMgYXJlCiAgICAjICAgIHJlbGF0ZWQuIFRoZSByZWZlcmVuY2UncyBgX19sZGdgIGlzIG9uZSBgTERHLkUuUzguQ09OU1RBTlQuU1lTYC4gVGhhdAogICAgIyAgICBpcyA0eCB0aGUgcmVhZCBpbnN0cnVjdGlvbnMgb24gdGhlIGhvdCBwYXRoIGFuZCBpdCBpcyBhIGNvc3Qgb2YgdGhlCiAgICAjICAgIG1hc2tlZCBmb3JtdWxhdGlvbiwgbm90IGEgY29kZWdlbiBmYWlsdXJlLiBTZWUgbm90ZXMvMTkgZ2FwIDcuCiAgICAjCiAgICAjIFRoZSBib3VuZHMgcHJlZGljYXRlIGlzIHdyaXR0ZW4gdGhlIG9idmlvdXMgc2lnbmVkIHdheS4gVGhlIHJlZmVyZW5jZQogICAgIyBmb2xkcyBlYWNoIGF4aXMgaW50byBvbmUgYCh1bnNpZ25lZCl2IDwgKHVuc2lnbmVkKXdgLCB3aGljaCBwdHhhcyB0dXJucwogICAgIyBpbnRvIGEgc2luZ2xlIGBJU0VUUC5HRS5VMzIuT1JgOyB0aGF0IGZvbGQgaXMgYSBzb3VyY2UtbGV2ZWwgZW5jb2RpbmcKICAgICMgdHJpY2ssIG5vdCBhIGRpZmZlcmVudCBwcmVkaWNhdGUsIGFuZCBpdCBpcyBsZWZ0IG91dCBoZXJlIG9uIHB1cnBvc2UgLS0KICAgICMgc2VlIG5vdGVzLzE5LCB3aGljaCBwcmljZXMgaXQgcmF0aGVyIHRoYW4gcXVpZXRseSB0YWtpbmcgaXQuCiAgICBiX2luZGV4ID0gVGVuc29yLmFyYW5nZShCLCBkdHlwZT1pMzIpLnJlc2hhcGUoQiwgMSwgMSwgMSkKCiAgICBoaXQgPSBUZW5zb3IuZnVsbCgoQiwgMSwgaGVpZ2h0LCB3aWR0aCksIEFJUiwgZHR5cGU9aTMyKQogICAgZm91bmQgPSBUZW5zb3IuZnVsbCgoQiwgMSwgMSwgMSksIEZhbHNlLCBkdHlwZT1kdHlwZXMuYm9vbCkKICAgIHN0ZXBzID0gVGVuc29yLnplcm9zKChCLCAxLCBoZWlnaHQsIHdpZHRoKSwgZHR5cGU9aTMyKQoKICAgIGZvciBzdGVwIGluIHJhbmdlKG1heF9zdGVwcyk6CiAgICAgICAgaWYgY291bnRfc3RlcHM6CiAgICAgICAgICAgICMgdGhlIGJvZHkgcnVucyBvbmNlIHBlciByYXkgcGVyIHN0ZXAgdW50aWwgaXQgaGl0cywgc28gdGhlIGNvdW50IGlzCiAgICAgICAgICAgICMgdGhlIDEtYmFzZWQgaW5kZXggb2YgdGhlIHN0ZXAgdGhhdCBoaXQsIG9yIG1heF9zdGVwcwogICAgICAgICAgICBzdGVwcyA9ICh+Zm91bmQpLndoZXJlKHN0ZXAgKyAxLCBzdGVwcykKCiAgICAgICAgaW5zaWRlID0gKCh2b3hlbF94ID49IDApICYgKHZveGVsX3ggPCBmbG9hdCh3eCkpICYKICAgICAgICAgICAgICAgICAgKHZveGVsX3kgPj0gMCkgJiAodm94ZWxfeSA8IGZsb2F0KHd5KSkgJgogICAgICAgICAgICAgICAgICAodm94ZWxfeiA+PSAwKSAmICh2b3hlbF96IDwgZmxvYXQod3opKSkKICAgICAgICBpeCA9IHZveGVsX3guY2FzdChpMzIpLmNsaXAoMCwgd3ggLSAxKQogICAgICAgIGl5ID0gdm94ZWxfeS5jYXN0KGkzMikuY2xpcCgwLCB3eSAtIDEpCiAgICAgICAgaXogPSB2b3hlbF96LmNhc3QoaTMyKS5jbGlwKDAsIHd6IC0gMSkKICAgICAgICBibG9jayA9IHZveGVsc1tiX2luZGV4LCBpeSwgaXgsIGl6XS5jYXN0KGkzMikKCiAgICAgICAgc29saWQgPSAoYmxvY2sgPj0gMCkgJiBpbnNpZGUgJiB+Zm91bmQKICAgICAgICBoaXQgPSBzb2xpZC53aGVyZShibG9jaywgaGl0KQogICAgICAgIGZvdW5kID0gZm91bmQgfCBzb2xpZAoKICAgICAgICAjIHRoZSBzdGVwIGl0c2VsZiwga2VybmVscy5jdTo2NzItNjc4LCBhcyB0aHJlZSBzZWxlY3RzCiAgICAgICAgc3RlcF90YWtlbl94ID0gKHRfbWF4X3ggPCB0X21heF95KSAmICh0X21heF94IDwgdF9tYXhfeikKICAgICAgICBzdGVwX3Rha2VuX3kgPSAodF9tYXhfeSA8IHRfbWF4X3opICYgfnN0ZXBfdGFrZW5feAogICAgICAgIHRfbWF4X3ggPSBzdGVwX3Rha2VuX3gud2hlcmUodF9tYXhfeCArIHRfZGVsdGFfeCwgdF9tYXhfeCkKICAgICAgICB0X21heF95ID0gc3RlcF90YWtlbl95LndoZXJlKHRfbWF4X3kgKyB0X2RlbHRhX3ksIHRfbWF4X3kpCiAgICAgICAgdF9tYXhfeiA9ICh+c3RlcF90YWtlbl94ICYgfnN0ZXBfdGFrZW5feSkud2hlcmUodF9tYXhfeiArIHRfZGVsdGFfeiwgdF9tYXhfeikKICAgICAgICB2b3hlbF94ID0gc3RlcF90YWtlbl94LndoZXJlKHZveGVsX3ggKyBzdGVwX3gsIHZveGVsX3gpCiAgICAgICAgdm94ZWxfeSA9IHN0ZXBfdGFrZW5feS53aGVyZSh2b3hlbF95ICsgc3RlcF95LCB2b3hlbF95KQogICAgICAgIHZveGVsX3ogPSAofnN0ZXBfdGFrZW5feCAmIH5zdGVwX3Rha2VuX3kpLndoZXJlKHZveGVsX3ogKyBzdGVwX3osIHZveGVsX3opCgogICAgIyAtLS0tIHRoZSBzdG9yZSwga2VybmVscy5jdTo2OTAtNjk5LCBhcyBhIHRhYmxlIGxvb2t1cAogICAgIwogICAgIyBgY29sb3VyX2x1dGAgaGFzIDE0IHJvd3M6IDAuLjEyIGFyZSB0aGUgYmxvY2tzLCByb3cgMTMgaXMgdGhlIHNreS4gQSByYXkKICAgICMgdGhhdCBoaXQgaG9sZHMgYSBibG9jayBpZCBpbiBbMCwgMTJdOyBhIHJheSB0aGF0IGRpZCBub3Qgc3RpbGwgaG9sZHMgQUlSCiAgICAjIGFuZCBpcyBtYXBwZWQgdG8gcm93IDEzLiBUaGUgcmVmZXJlbmNlIHJlYWNoZXMgdGhlIHNreSBieSBhICpicmFuY2gqCiAgICAjIChgaWYgKGhpdF9ibG9jayA+PSAwKSAuLi4gZWxzZSAuLi5gKTsgaGVyZSBib3RoIGFyZSB0aGUgc2FtZSBnYXRoZXIuCiAgICAjIFRoYXQgaXMgYSBjaGFuZ2Ugb2YgZXhwcmVzc2lvbiwgbm90IG9mIHNjaGVkdWxlIC0tIG5vIGtub2IsIG5vIGZ1c2lvbgogICAgIyBoaW50LCBubyB0aWxlIHNpemUuCiAgICBsdXQgPSBjb2xvdXJfbHV0KCkgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAgICAjIFsxNCwgM10gdWludDgKICAgIGNvbG91cl9yb3cgPSAoaGl0ID49IDApLndoZXJlKGhpdCwgMTMpICAgICAgICAgICAgICAgICAgICMgW0IsIDEsIEgsIFddIGludDMyCiAgICBvYnMgPSBsdXRbY29sb3VyX3Jvd10ucmVzaGFwZShCLCBoZWlnaHQsIHdpZHRoLCAzKSAgICAgICAjIFtCLCBILCBXLCAzXSB1aW50OAogICAgcmV0dXJuIChvYnMsIHN0ZXBzKSBpZiBjb3VudF9zdGVwcyBlbHNlIG9icy5jb250aWd1b3VzKCkK", "fe85581f80ed7fdba57398a6c04fe744a7fd59917d99b132ea0eb2dcfa457952"], ["minecu_tg/cpu_reference.py", "IiIiQSBmbG9hdDY0IENQVSByZWZlcmVuY2Ugb2YgdGhlIHNhbWUgYWxnb3JpdGhtLgoKV0hZIEEgU0VQQVJBVEUgUkVGRVJFTkNFLCBJTlNURUFEIE9GIENPTVBBUklORyBUTyBUSEUgR1BVLiBUaGUgdHdvIHJlbmRlcmVycwphcmUgbm90IGJpdC1jb21wYXJhYmxlIGFuZCBpdCBpcyBub3Qgd29ydGggcHJldGVuZGluZyBvdGhlcndpc2U6CgogICogdGhlIHJlZmVyZW5jZSBydW5zIGAtLXVzZV9mYXN0X21hdGhgLCBzbyBgc2luZmAvYGNvc2ZgL2B0YW5mYC9gcnNxcnRmYCBhbmQKICAgIHRoZSBkaXZpc2lvbnMgYXJlIGFwcHJveGltYXRlLCBhbmQgYF9fc2luY29zZmAgaXMgdGhlIGZhc3QgaW50cmluc2ljOwogICogVGlueUdyYWQncyBDVURBIGJhY2tlbmQgbG93ZXJzIGBzaW5gL2Bjb3NgL2B0YW5gL2Byc3FydGAgdG8gaXRzIG93bgogICAgc2VxdWVuY2UsIGFuZCBpdHMgYC9gIGlzIGEgcGxhaW4gZGl2aWRlOwogICogYSBEREEgc3RlcCBpcyBjaG9zZW4gYnkgYHRfbWF4X3ggPCB0X21heF95YCwgc28gYSBvbmUtdWxwIGRpZmZlcmVuY2UgaW4gYQogICAgYHRfbWF4YCBjYW4gbW92ZSBhIHJheSB0byBhIGRpZmZlcmVudCB2b3hlbCwgYW5kIHRoZW4gdGhlIGltYWdlIGRpZmZlcnMgYnkKICAgIGEgd2hvbGUgYmxvY2sgY29sb3VyLgoKU28gdGhlIGNvbnRyYWN0IHRoaXMgbW9kdWxlIGVuZm9yY2VzIGlzOiAqKnRoZSB0ZW5zb3IgcmVuZGVyIGFncmVlcyB3aXRoIGFuCmluZGVwZW5kZW50IGZsb2F0NjQgdHJhbnNjcmlwdGlvbiBvZiB0aGUgc2FtZSBhbGdvcml0aG0uKiogVGhlIGZsb2F0NjQKdHJhbnNjcmlwdGlvbiBpcyB3cml0dGVuIGZyb20gYG5vdGVzLzE3LW1pbmVjdS1zcGVjLm1kYCBhbmQgZnJvbSB0aGUgcmVmZXJlbmNlJ3MKc291cmNlLCBhbmQgaXQgaXMgY2hlY2tlZCBhZ2FpbnN0IHRoZSByZWZlcmVuY2UncyBvd24gQ1BVIG1vZGVsCihgbWluZWN1X3JlZi9iZW5jaC9kZGFfbW9kZWwucHlgLCBhbiBpbmRlcGVuZGVudCB0aGlyZCB0cmFuc2NyaXB0aW9uKSBzbyB0aGF0IGEKYnVnIGluIHRoaXMgZmlsZSBjYW5ub3Qgc2lsZW50bHkgYmVjb21lIHRoZSBkZWZpbml0aW9uIG9mIGNvcnJlY3QuCgpUaGUgdHdvIG91dHB1dC1sZXZlbCBjbGFpbXMgYHZlcmlmeS5weWAgbWFrZXMgYXJlIHRoZXJlZm9yZToKCiAgMS4gdGhlIFRpbnlHcmFkIGltYWdlIGVxdWFscyB0aGlzIGZsb2F0NjQgaW1hZ2UsIGV4YWN0bHksIG9uIHVpbnQ4IGNvbG91cnM7CiAgMi4gdGhlIG1lYW4gRERBIHN0ZXBzIHBlciBwaXhlbCBhZ3JlZSB0byAzIGRlY2ltYWwgcGxhY2VzIHdpdGggdGhlCiAgICAgcmVmZXJlbmNlJ3Mgb3duIG1lYXN1cmVkIDI2LjQxNiBhdCB0aGUgb24tc3VyZmFjZSBjb25maWd1cmF0aW9uLgoKQ2xhaW0gMiBpcyB0aGUgb25lIHRoYXQgYWN0dWFsbHkgcGlucyB0aGUgKndvcmtsb2FkKiwgYmVjYXVzZSBpdCBpcyBhCmRpc2NyZXRlLCB0aWUtc2Vuc2l0aXZlIHF1YW50aXR5OiBpZiB0aGUgdHdvIGltcGxlbWVudGF0aW9ucyBtYWRlIGRpZmZlcmVudAp0cmF2ZXJzYWwgZGVjaXNpb25zLCB0aGUgc3RlcCBjb3VudCB3b3VsZCBtb3ZlLgoiIiIKCmltcG9ydCBtYXRoCgppbXBvcnQgbnVtcHkgYXMgbnAKCmZyb20gLmJsb2NrcyBpbXBvcnQgQkxPQ0tfQ09MT1JTX1U4LCBTS1lfQ09MT1JfVTgKCkFJUiA9IC0xCiMgVGhlIDE0LXJvdyB0YWJsZTogMC4uMTIgYXJlIHRoZSBibG9ja3MsIHJvdyAxMyBpcyB0aGUgc2t5LiBTYW1lIHRhYmxlIHRoZQojIHRlbnNvciB2ZXJzaW9uIGdhdGhlcnMgZnJvbSwgc28gdGhlIGNvbXBhcmlzb24gaXNvbGF0ZXMgdGhlIEREQSBhbmQgbm90IHRoZQojIHBhbGV0dGUuCkxVVCA9IG5wLmFycmF5KEJMT0NLX0NPTE9SU19VOCArIFtTS1lfQ09MT1JfVThdLCBkdHlwZT1ucC51aW50OCkKCgpkZWYgY2FtZXJhX2Jhc2lzKGNhbWVyYXM6IG5wLm5kYXJyYXksIGZvdl9kZWdyZWVzOiBmbG9hdCwgd2lkdGg6IGludCwgaGVpZ2h0OiBpbnQpIC0+IG5wLm5kYXJyYXk6CiAgICAiIiJgcmVuZGVyLmNhbWVyYV9iYXNpc2AsIGluIGZsb2F0NjQsIGZvciBvbmUgY2FtZXJhLgoKICAgIFJldHVybnMgdGhlIHNhbWUgMTQgbnVtYmVycyB0aGUga2VybmVsIHN0b3JlcywgaW4gdGhlIHNhbWUgb3JkZXIsIHNvIHRoZQogICAgY29tcGFyaXNvbiBpcyBub3Qgc2Vuc2l0aXZlIHRvIHRoZSBwYWNraW5nLgogICAgIiIiCiAgICBjYW1feCwgY2FtX3ksIGNhbV96LCB5YXcsIHBpdGNoID0gKGZsb2F0KG5wLnJlc2hhcGUoY2FtZXJhcywgLTEpW2ldKSBmb3IgaSBpbiByYW5nZSg1KSkKICAgIHNpbl95YXcsIGNvc195YXcgPSBtYXRoLnNpbih5YXcpLCBtYXRoLmNvcyh5YXcpCiAgICBzaW5fcGl0Y2gsIGNvc19waXRjaCA9IG1hdGguc2luKHBpdGNoKSwgbWF0aC5jb3MocGl0Y2gpCiAgICBoYWxmX2ZvdiA9IG1hdGgudGFuKGZvdl9kZWdyZWVzICogMC4wMTc0NTMyOSAqIDAuNSkKICAgIGFzcGVjdCA9IHdpZHRoIC8gaGVpZ2h0CiAgICByZXR1cm4gbnAuYXJyYXkoWwogICAgICAgIGNhbV94LCBjYW1feSArIDEuNiwgY2FtX3osCiAgICAgICAgc2luX3lhdyAqIGNvc19waXRjaCwgc2luX3BpdGNoLCBjb3NfeWF3ICogY29zX3BpdGNoLAogICAgICAgIGNvc195YXcsIC1zaW5feWF3LAogICAgICAgIC1zaW5feWF3ICogc2luX3BpdGNoLCBjb3NfcGl0Y2gsIC1jb3NfeWF3ICogc2luX3BpdGNoLAogICAgICAgIGhhbGZfZm92ICogYXNwZWN0LCBoYWxmX2ZvdiwKICAgIF0sIGR0eXBlPW5wLmZsb2F0NjQpCgoKZGVmIHJlbmRlcl9vbmUodm94ZWxzOiBucC5uZGFycmF5LCBiYXNpczogbnAubmRhcnJheSwgd2lkdGg6IGludCwgaGVpZ2h0OiBpbnQsCiAgICAgICAgICAgICAgIG1heF9zdGVwczogaW50KToKICAgICIiIk9uZSBlbnYuIHZveGVscyBpcyBbV1ksIFdYLCBXWl0gaW50ODsgYmFzaXMgaXMgdGhlIDE0LXZlY3Rvci4KCiAgICBSZXR1cm5zIChpbWFnZSBbSCwgVywgM10gdWludDgsIHN0ZXBzIFtILCBXXSBpbnQzMiksIHRoZSBzdGVwIGNvdW50IGJlaW5nCiAgICB0aGUgbnVtYmVyIG9mIHRpbWVzIHRoZSBEREEgYm9keSByYW4sIHdoaWNoIGlzIHRoZSBxdWFudGl0eSB0aGUgcmVmZXJlbmNlJ3MKICAgIGRldmljZS1zaWRlIHByb2JlIHJlcG9ydHMuCiAgICAiIiIKICAgIHd5LCB3eCwgd3ogPSB2b3hlbHMuc2hhcGUKICAgIG91dCA9IG5wLmVtcHR5KChoZWlnaHQsIHdpZHRoLCAzKSwgZHR5cGU9bnAudWludDgpCiAgICBzdGVwcyA9IG5wLnplcm9zKChoZWlnaHQsIHdpZHRoKSwgZHR5cGU9bnAuaW50MzIpCgogICAgZXllX3gsIGV5ZV95LCBleWVfeiA9IGJhc2lzWzBdLCBiYXNpc1sxXSwgYmFzaXNbMl0KICAgIGZ3ZF94LCBmd2RfeSwgZndkX3ogPSBiYXNpc1szXSwgYmFzaXNbNF0sIGJhc2lzWzVdCiAgICByaWdodF94LCByaWdodF96ID0gYmFzaXNbNl0sIGJhc2lzWzddCiAgICB1cF94LCB1cF95LCB1cF96ID0gYmFzaXNbOF0sIGJhc2lzWzldLCBiYXNpc1sxMF0KICAgIHNjYWxlX3UsIHNjYWxlX3YgPSBiYXNpc1sxMV0sIGJhc2lzWzEyXQoKICAgIGZvciBweSBpbiByYW5nZShoZWlnaHQpOgogICAgICAgIGZvciBweCBpbiByYW5nZSh3aWR0aCk6CiAgICAgICAgICAgICMga2VybmVscy5jdTo2MTktNjI5CiAgICAgICAgICAgIHUgPSAoMi4wICogKHB4ICsgMC41KSAvIHdpZHRoIC0gMS4wKSAqIHNjYWxlX3UKICAgICAgICAgICAgdiA9ICgxLjAgLSAyLjAgKiAocHkgKyAwLjUpIC8gaGVpZ2h0KSAqIHNjYWxlX3YKICAgICAgICAgICAgcmF5X3ggPSBmd2RfeCArIHUgKiByaWdodF94ICsgdiAqIHVwX3gKICAgICAgICAgICAgcmF5X3kgPSBmd2RfeSArIHYgKiB1cF95CiAgICAgICAgICAgIHJheV96ID0gZndkX3ogKyB1ICogcmlnaHRfeiArIHYgKiB1cF96CiAgICAgICAgICAgIHJsID0gMS4wIC8gbWF0aC5zcXJ0KHJheV94ICogcmF5X3ggKyByYXlfeSAqIHJheV95ICsgcmF5X3ogKiByYXlfeikKICAgICAgICAgICAgcmF5X3gsIHJheV95LCByYXlfeiA9IHJheV94ICogcmwsIHJheV95ICogcmwsIHJheV96ICogcmwKCiAgICAgICAgICAgICMga2VybmVscy5jdTo2MzgtNjU5CiAgICAgICAgICAgIHZ4LCB2eSwgdnogPSBpbnQobWF0aC5mbG9vcihleWVfeCkpLCBpbnQobWF0aC5mbG9vcihleWVfeSkpLCBpbnQobWF0aC5mbG9vcihleWVfeikpCiAgICAgICAgICAgIHN4ID0gMSBpZiByYXlfeCA+PSAwIGVsc2UgLTEKICAgICAgICAgICAgc3kgPSAxIGlmIHJheV95ID49IDAgZWxzZSAtMQogICAgICAgICAgICBzeiA9IDEgaWYgcmF5X3ogPj0gMCBlbHNlIC0xCiAgICAgICAgICAgIGl4ID0gMS4wIC8gcmF5X3ggaWYgcmF5X3ggIT0gMCBlbHNlIDFlMzAKICAgICAgICAgICAgaXkgPSAxLjAgLyByYXlfeSBpZiByYXlfeSAhPSAwIGVsc2UgMWUzMAogICAgICAgICAgICBpeiA9IDEuMCAvIHJheV96IGlmIHJheV96ICE9IDAgZWxzZSAxZTMwCiAgICAgICAgICAgIGR0eCwgZHR5LCBkdHogPSBhYnMoaXgpLCBhYnMoaXkpLCBhYnMoaXopCiAgICAgICAgICAgIHRteCA9ICh2eCArIDEgLSBleWVfeCkgKiBpeCBpZiByYXlfeCA+PSAwIGVsc2UgKHZ4IC0gZXllX3gpICogaXgKICAgICAgICAgICAgdG15ID0gKHZ5ICsgMSAtIGV5ZV95KSAqIGl5IGlmIHJheV95ID49IDAgZWxzZSAodnkgLSBleWVfeSkgKiBpeQogICAgICAgICAgICB0bXogPSAodnogKyAxIC0gZXllX3opICogaXogaWYgcmF5X3ogPj0gMCBlbHNlICh2eiAtIGV5ZV96KSAqIGl6CgogICAgICAgICAgICAjIGtlcm5lbHMuY3U6NjY0LTY4Ny4gYG5gIGNvdW50cyBib2R5IGV4ZWN1dGlvbnMsIHdoaWNoIGlzIHdoYXQgdGhlCiAgICAgICAgICAgICMgcmVmZXJlbmNlJ3MgcHJvYmUgcmVwb3J0cyBhcyAiRERBIHN0ZXBzL3BpeGVsIi4KICAgICAgICAgICAgaGl0ID0gQUlSCiAgICAgICAgICAgIG4gPSAwCiAgICAgICAgICAgIGZvciBfIGluIHJhbmdlKG1heF9zdGVwcyk6CiAgICAgICAgICAgICAgICBuICs9IDEKICAgICAgICAgICAgICAgIGlmIDAgPD0gdnggPCB3eCBhbmQgMCA8PSB2eSA8IHd5IGFuZCAwIDw9IHZ6IDwgd3o6CiAgICAgICAgICAgICAgICAgICAgYmxrID0gaW50KHZveGVsc1t2eSwgdngsIHZ6XSkKICAgICAgICAgICAgICAgICAgICBpZiBibGsgPj0gMDoKICAgICAgICAgICAgICAgICAgICAgICAgaGl0ID0gYmxrCiAgICAgICAgICAgICAgICAgICAgICAgIGJyZWFrCiAgICAgICAgICAgICAgICBpZiB0bXggPCB0bXkgYW5kIHRteCA8IHRtejoKICAgICAgICAgICAgICAgICAgICB0bXggKz0gZHR4OyB2eCArPSBzeAogICAgICAgICAgICAgICAgZWxpZiB0bXkgPCB0bXo6CiAgICAgICAgICAgICAgICAgICAgdG15ICs9IGR0eTsgdnkgKz0gc3kKICAgICAgICAgICAgICAgIGVsc2U6CiAgICAgICAgICAgICAgICAgICAgdG16ICs9IGR0ejsgdnogKz0gc3oKICAgICAgICAgICAgc3RlcHNbcHksIHB4XSA9IG4KICAgICAgICAgICAgb3V0W3B5LCBweF0gPSBMVVRbMTMgaWYgaGl0IDwgMCBlbHNlIGhpdF0KICAgIHJldHVybiBvdXQsIHN0ZXBzCg==", "d3d6356552db2be483430751924970d288cb1c128a56709aa074d7e18a908108"], ["minecu_tg/bench.py", "IiIiVGltaW5nIGFuZCBtZW1vcnksIG9uIGEgcmVhbCBkZXZpY2UuCgpXSEFUIElTIFRJTUVELCBBTkQgV0hZIElUIElTIE5PVCBUSEUgV0hPTEUgU1RFUAotLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0KVGhlIGhhbmQtd3JpdHRlbiBiYXNlbGluZSdzIG51bWJlciBpcyBhIHdob2xlICpzdGVwKjogc2V2ZW4gb3IgbmluZSBsYXVuY2hlcywKZml2ZSBvZiB0aGVtIHN0YXRlIGtlcm5lbHMsIGFuZCB0aGUgbGFzdCBvbmUgdGhlIHJlbmRlci4gVGlueUdyYWQsIGdpdmVuIHRoZQpyZW5kZXIgYWxvbmUsIGVtaXRzIGEgZ3JhcGggb2Yga2VybmVscy4gVGhvc2UgYXJlIGRpZmZlcmVudCBvYmplY3RzLCBhbmQKYG5vdGVzLzE3YCDCpzguMiBzYXlzIHNvIGJlZm9yZSBhbnkgbnVtYmVyIGlzIHRha2VuLiBUaGlzIG1vZHVsZSB0aGVyZWZvcmUgdGltZXMKdGhyZWUgc2VwYXJhdGUgdGhpbmdzIGFuZCBsYWJlbHMgYWxsIHRocmVlOgoKICAqIGByZW5kZXJgICAgICAgLS0gdGhlIFRpbnlHcmFkIGdyYXBoIGZvciB0aGUgb2JzZXJ2YXRpb24sIGFuZCBub3RoaW5nIGVsc2UuCiAgICAgICAgICAgICAgICAgICAgIFRoaXMgaXMgdGhlIHBlci1vcCBudW1iZXIsIGNvbXBhcmFibGUgdG8gdGhlIHJlZmVyZW5jZSdzCiAgICAgICAgICAgICAgICAgICAgIGByZW5kZXJfa2VybmVsX3VpbnQ4X3ByZWJhc2lzYCBwZXIta2VybmVsIG1lZGlhbi4KICAqIGByZW5kZXIrZDJoYCAgLS0gdGhlIHNhbWUgcGx1cyBhIHJlYWQtYmFjaywgc28gdGhlIGNvc3Qgb2YgZ2V0dGluZyB0aGUKICAgICAgICAgICAgICAgICAgICAgYW5zd2VyIG9mZiB0aGUgZGV2aWNlIGlzIHZpc2libGUgcmF0aGVyIHRoYW4gYXNzdW1lZCBhd2F5LgogICogYHdob2xlIGdyYXBoYCAtLSB0aGUgd2hvbGUgYG9icyA9IHJlbmRlciguLi4pYCBpbmNsdWRpbmcgdGhlIGJhc2lzIGtlcm5lbCwKICAgICAgICAgICAgICAgICAgICAgd2hpY2ggaXMgdGhlIG9iamVjdCBtb3N0IGNvbXBhcmFibGUgdG8gdGhlIHJlZmVyZW5jZSdzCiAgICAgICAgICAgICAgICAgICAgIHdob2xlIHN0ZXAuIEl0IGlzIHN0aWxsIG5vdCBhIHdob2xlIHN0ZXAsIGJlY2F1c2UgdGhlCiAgICAgICAgICAgICAgICAgICAgIHJlZmVyZW5jZSdzIHN0ZXAgYWxzbyBkb2VzIGRlY29kZS9waHlzaWNzL3JheWNhc3QvcmVzZXQsCiAgICAgICAgICAgICAgICAgICAgIGFuZCB0aGF0IGRpZmZlcmVuY2UgaXMgbmFtZWQgcmF0aGVyIHRoYW4gaGlkZGVuLgoKV0hZIE1FTU9SWSBJUyBNRUFTVVJFRCBBTkQgTk9UIERFUklWRUQKLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tLS0tCkF0IGBtYXhfc3RlcHM9MzJgIHRoZSBnZW5lcmF0ZWQgc2NoZWR1bGUgd3JpdGVzIHRoZSBEREEgc3RhdGUgdG8gZ2xvYmFsIG1lbW9yeQpvbmNlIHBlciBzdGVwIGFuZCB0aGUgZmluYWwga2VybmVsIHRha2VzIGFsbCBvZiB0aG9zZSBidWZmZXJzIGFzIGlucHV0cy4gVGhhdAptYWtlcyB0aGUgcGVhayBmb290cHJpbnQgYSBmdW5jdGlvbiBvZiBgbWF4X3N0ZXBzYCBhbmQgb2YgdGhlIHBpeGVsIGNvdW50LCBhbmQKYXQgdGhlIHByb2plY3QncyBoZWFkbGluZSBzaGFwZSAoYmF0Y2ggMzI3NjgsIDMyeDMyKSB0aGUgYXJpdGhtZXRpYyBzYXlzIHRoZQpncmFwaCBjYW5ub3QgYmUgYWxsb2NhdGVkIGF0IGFsbC4gQW4gYXJpdGhtZXRpYyBjb25zZXF1ZW5jZSBpcyBub3QgYQptZWFzdXJlbWVudCwgc28gdGhlIGZvb3RwcmludCBpcyBtZWFzdXJlZCB3aXRoIHRoZSBhbGxvY2F0b3IncyBvd24gY291bnRlcnMgYW5kCnRoZSBydW4gcmVwb3J0cyB3aGF0IGl0IGFjdHVhbGx5IGFsbG9jYXRlZCwgaW5jbHVkaW5nIHRoZSBzaGFwZXMgd2hlcmUgaXQKZmFpbGVkLgoiIiIKCmZyb20gX19mdXR1cmVfXyBpbXBvcnQgYW5ub3RhdGlvbnMKCmltcG9ydCBnYwppbXBvcnQgc3RhdGlzdGljcwppbXBvcnQgdGltZQoKCmRlZiBfc3luYyhkZXYpOgogICAgZnJvbSB0aW55Z3JhZCBpbXBvcnQgRGV2aWNlCiAgICBkID0gRGV2aWNlW2Rldl0KICAgIGlmIGhhc2F0dHIoZCwgInN5bmNocm9uaXplIik6CiAgICAgICAgZC5zeW5jaHJvbml6ZSgpCgoKY2xhc3MgTWVtUHJvYmU6CiAgICAiIiJUcmFjayB3aGF0IHRoZSBhbGxvY2F0b3IgYWN0dWFsbHkgaGFuZHMgb3V0LCBieSB3cmFwcGluZyBgYWxsb2NgL2BmcmVlYC4KCiAgICBUaGlzIHZlcnNpb24gb2YgdGlueWdyYWQgaGFzIG5vIGBtYXhfbWVtb3J5X2FsbG9jYXRlZGAsIHNvIHRoZSBmb290cHJpbnQgaXMKICAgIG1lYXN1cmVkIGJ5IGluc3RydW1lbnRpbmcgdGhlIG9uZSBwbGFjZSBieXRlcyBhcmUgcmVxdWVzdGVkOgogICAgYEFsbG9jYXRvci5hbGxvY2AgLyBgQWxsb2NhdG9yLmZyZWVgIChkZXZpY2UucHkpLiBgYWxsb2NgIHJlY29yZHMgdGhlCiAgICByZXF1ZXN0ZWQgc2l6ZSBhbmQgYWRkcyBpdCB0byBhIGxpdmUgdG90YWw7IGBmcmVlYCBzdWJ0cmFjdHMuIGBwZWFrYCBpcyB0aGUKICAgIGhpZ2gtd2F0ZXIgbWFyayBvZiB0aGF0IHRvdGFsLgoKICAgIFRoaXMgaXMgYSBtZWFzdXJlbWVudCBvZiB0aGUgcmVhbCBhbGxvY2F0b3IsIG5vdCBhbiBlc3RpbWF0ZSBmcm9tIHRoZQogICAgZ2VuZXJhdGVkIHNvdXJjZSwgYW5kIHRoZSB0d28gYXJlIHJlcG9ydGVkIHNlcGFyYXRlbHkgaW4gdGhlIG5vdGVzIGJlY2F1c2UKICAgIHRoZXkgYW5zd2VyIGRpZmZlcmVudCBxdWVzdGlvbnM6IHRoZSBhbGxvY2F0b3IncyBwZWFrIGlzIHdoYXQgdGhlIGNhcmQgaGFkCiAgICB0byBoYXZlLCBhbmQgdGhlIHNvdXJjZSdzIGFyZ3VtZW50IGxpc3QgaXMgd2hhdCB0aGUgc2NoZWR1bGUgc3RydWN0dXJhbGx5CiAgICByZXF1aXJlcy4KCiAgICBgb29tYCBpcyB0aGUgZXhjZXB0aW9uIHRleHQgd2hlbiBhbiBhbGxvY2F0aW9uIGZhaWxzLCB3aGljaCBhdCB0aGUgaGVhZGxpbmUKICAgIHNoYXBlIGlzIHRoZSBmaW5kaW5nIHJhdGhlciB0aGFuIGEgZmFpbHVyZSB0byBoaWRlLgogICAgIiIiCgogICAgZGVmIF9faW5pdF9fKHNlbGYsIGRldjogc3RyKToKICAgICAgICBmcm9tIHRpbnlncmFkIGltcG9ydCBEZXZpY2UKICAgICAgICBmcm9tIHRpbnlncmFkLmRldmljZSBpbXBvcnQgQWxsb2NhdG9yCiAgICAgICAgc2VsZi5kZXYgPSBkZXYKICAgICAgICBzZWxmLmxpdmUgPSAwCiAgICAgICAgc2VsZi5wZWFrID0gMAogICAgICAgIHNlbGYudG90YWwgPSAwCiAgICAgICAgc2VsZi5uX2FsbG9jcyA9IDAKICAgICAgICBzZWxmLm9vbTogc3RyIHwgTm9uZSA9IE5vbmUKICAgICAgICBzZWxmLmQgPSBEZXZpY2VbZGV2XQogICAgICAgIHNlbGYuX2EsIHNlbGYuX2YgPSBBbGxvY2F0b3IuYWxsb2MsIEFsbG9jYXRvci5mcmVlCiAgICAgICAgcHJvYmUgPSBzZWxmICAgICAgICAgICAgICAgICAgICAjIHRoZSBjbG9zdXJlcyB0YWtlIHRoZSBhbGxvY2F0b3IgYXMgYGFgCgogICAgICAgIGRlZiBhbGxvYyhhLCBzaXplLCBvcHRpb25zPU5vbmUpOgogICAgICAgICAgICByID0gcHJvYmUuX2EoYSwgc2l6ZSwgb3B0aW9ucykKICAgICAgICAgICAgcHJvYmUubl9hbGxvY3MgKz0gMQogICAgICAgICAgICBwcm9iZS50b3RhbCArPSBpbnQoc2l6ZSkKICAgICAgICAgICAgcHJvYmUubGl2ZSArPSBpbnQoc2l6ZSkKICAgICAgICAgICAgcHJvYmUucGVhayA9IG1heChwcm9iZS5wZWFrLCBwcm9iZS5saXZlKQogICAgICAgICAgICByZXR1cm4gcgoKICAgICAgICBkZWYgZnJlZShhLCBvcGFxdWUsIHNpemUsIG9wdGlvbnM9Tm9uZSk6CiAgICAgICAgICAgIHByb2JlLmxpdmUgLT0gaW50KHNpemUpCiAgICAgICAgICAgIHJldHVybiBwcm9iZS5fZihhLCBvcGFxdWUsIHNpemUsIG9wdGlvbnMpCgogICAgICAgIGRlZiBhbGxvY19vb20oYSwgc2l6ZSwgb3B0aW9ucz1Ob25lKToKICAgICAgICAgICAgdHJ5OgogICAgICAgICAgICAgICAgcmV0dXJuIGFsbG9jKGEsIHNpemUsIG9wdGlvbnMpCiAgICAgICAgICAgIGV4Y2VwdCBNZW1vcnlFcnJvciBhcyBlOgogICAgICAgICAgICAgICAgcHJvYmUub29tID0gZiJ7ZX0iWzozMDBdCiAgICAgICAgICAgICAgICByYWlzZQoKICAgICAgICBBbGxvY2F0b3IuYWxsb2MsIEFsbG9jYXRvci5mcmVlID0gYWxsb2Nfb29tLCBmcmVlCgogICAgZGVmIHJlc2V0KHNlbGYpOgogICAgICAgIHNlbGYubGl2ZSA9IHNlbGYucGVhayA9IHNlbGYudG90YWwgPSBzZWxmLm5fYWxsb2NzID0gMAogICAgICAgIHNlbGYub29tID0gTm9uZQoKICAgIGRlZiByZXN0b3JlKHNlbGYpOgogICAgICAgIGZyb20gdGlueWdyYWQuZGV2aWNlIGltcG9ydCBBbGxvY2F0b3IKICAgICAgICBBbGxvY2F0b3IuYWxsb2MsIEFsbG9jYXRvci5mcmVlID0gc2VsZi5fYSwgc2VsZi5fZgoKICAgIGRlZiBfX2VudGVyX18oc2VsZik6CiAgICAgICAgcmV0dXJuIHNlbGYKCiAgICBkZWYgX19leGl0X18oc2VsZiwgKmEpOgogICAgICAgIHNlbGYucmVzdG9yZSgpCiAgICAgICAgcmV0dXJuIEZhbHNlCgogICAgZGVmIGFzX2RpY3Qoc2VsZikgLT4gZGljdDoKICAgICAgICByZXR1cm4geyJwZWFrX2xpdmVfYnl0ZXMiOiBzZWxmLnBlYWssICJ0b3RhbF9hbGxvY2F0ZWRfYnl0ZXMiOiBzZWxmLnRvdGFsLAogICAgICAgICAgICAgICAgIm5fYWxsb2NzIjogc2VsZi5uX2FsbG9jcywgImxpdmVfYXRfZW5kX2J5dGVzIjogc2VsZi5saXZlLAogICAgICAgICAgICAgICAgIm9vbSI6IHNlbGYub29tfQoKCmRlZiB0aW1lX2NhbGwoZm4sIGRldiwgaXRlcnM6IGludCwgd2FybXVwOiBpbnQgPSA1KSAtPiBkaWN0OgogICAgIiIiTWVkaWFuIC8gbWluIC8gbWF4IC8gc3RkIG9mIGBmbigpYCwgaW4gbWlsbGlzZWNvbmRzLCB3YWxsIGNsb2NrLgoKICAgIFRoaXMgaXMgdGhlICplbmQgdG8gZW5kKiBudW1iZXI6IHdoYXRldmVyIGBmbmAgZG9lcywgaG9zdCBzaWRlIGFuZCBkZXZpY2UKICAgIHNpZGUsIHBsdXMgb25lIHN5bmNocm9uaXphdGlvbi4gQm90aCBoYWx2ZXMgYXJlIHJlYWwgY29zdHMgYW5kIG5laXRoZXIgaXMKICAgIGRpdmlkZWQgb3V0LCBiZWNhdXNlIHRoZSBwb2ludCBpcyB3aGF0IGEgY2FsbGVyIHBheXMuIGBkZXZpY2VfdGltZV9tc2AgaXMKICAgIHRoZSBkZXZpY2Utb25seSBjb3VudGVycGFydCwgbWVhc3VyZWQgc2VwYXJhdGVseS4KICAgICIiIgogICAgZm9yIF8gaW4gcmFuZ2Uod2FybXVwKToKICAgICAgICBmbigpCiAgICBfc3luYyhkZXYpCiAgICB0cyA9IFtdCiAgICBmb3IgXyBpbiByYW5nZShpdGVycyk6CiAgICAgICAgdDAgPSB0aW1lLnBlcmZfY291bnRlcigpCiAgICAgICAgZm4oKQogICAgICAgIF9zeW5jKGRldikKICAgICAgICB0cy5hcHBlbmQoKHRpbWUucGVyZl9jb3VudGVyKCkgLSB0MCkgKiAxZTMpCiAgICB0cy5zb3J0KCkKICAgIHJldHVybiB7CiAgICAgICAgIml0ZXJzIjogaXRlcnMsCiAgICAgICAgIm1lZGlhbl9tcyI6IHN0YXRpc3RpY3MubWVkaWFuKHRzKSwKICAgICAgICAibWluX21zIjogdHNbMF0sCiAgICAgICAgIm1heF9tcyI6IHRzWy0xXSwKICAgICAgICAibWVhbl9tcyI6IHN0YXRpc3RpY3MuZm1lYW4odHMpLAogICAgICAgICJzdGRfbXMiOiBzdGF0aXN0aWNzLnN0ZGV2KHRzKSBpZiBsZW4odHMpID4gMSBlbHNlIDAuMCwKICAgIH0KCgpkZWYgZGV2aWNlX3RpbWVfbXModGVuc29yLCBpdGVyczogaW50ID0gNSkgLT4gZGljdCB8IE5vbmU6CiAgICAiIiJEZXZpY2Utb25seSB0aW1lIGZvciBhIHJlYWxpemVkIHRlbnNvciwgdmlhIHRpbnlncmFkJ3Mgb3duIGB0aW1lX2NhbGxgLgoKICAgIGB0aW55Z3JhZC9lbmdpbmUvcmVhbGl6ZS5weTp0aW1lX2NhbGxgIGJ1aWxkcyBhbiBFeGVjQ29udGV4dCB3aXRoCiAgICBgdXBkYXRlX3N0YXRzPUZhbHNlLCB3YWl0PVRydWVgIGFuZCByZS1leGVjdXRlcyBhIHNpbmdsZSBDQUxMLCByZXR1cm5pbmcKICAgIHRoZSBkZXZpY2UtbWVhc3VyZWQgZHVyYXRpb24gb2YgdGhlIGxhcmdlc3Qga2VybmVsIGluIHRoZSBncm91cC4gVGhhdCBpcwogICAgdGhlIHNhbWUgcXVhbnRpdHkgYGJlbmNoLmN1YCBnZXRzIGZyb20gYGN1ZGFFdmVudGAgYXJvdW5kIGEgbGF1bmNoLCBhbmQgaXQKICAgIGlzIHRoZSByaWdodCBpbnN0cnVtZW50IGhlcmU6IHdhbGwgY2xvY2sgYXJvdW5kIGEgUHl0aG9uLWxldmVsIGNhbGwgaXMKICAgIGRvbWluYXRlZCBieSB0cmFjaW5nLCBhbmQgdHJhY2luZyBpcyBub3QgdGhlIHJlbmRlci4KCiAgICBUd28gY2F2ZWF0cywgYm90aCBjYXJyaWVkIGludG8gdGhlIG5vdGVzIHJhdGhlciB0aGFuIGhpZGRlbjoKCiAgICAgICogaXQgcmUtdXNlcyB0aGUgKmNvbXBpbGVkKiBwcm9ncmFtLCBzbyB0aGlzIG1lYXN1cmVzIHJlLWV4ZWN1dGlvbiwgbm90CiAgICAgICAgdGhlIGZpcnN0IGNhbGwncyBjb21waWxlLiBDb21waWxhdGlvbiBpcyByZXBvcnRlZCBzZXBhcmF0ZWx5ICh0aGUKICAgICAgICBjZW5zdXMncyBgbmV3X2Rpc3RpbmN0X3Byb2dyYW1zYCBhbmQgdGhlIHdhbGwtY2xvY2sgYnVpbGQtb25seSBmaWd1cmUpLgogICAgICAqIGl0IHRha2VzIHRoZSBtYXggb3ZlciB0aGUga2VybmVscyBpbiB0aGUgZ3JvdXAsIHNvIGZvciBhIGdyYXBoIGl0IGlzIGEKICAgICAgICBwZXIta2VybmVsIG51bWJlciwgbm90IGEgd2hvbGUtZ3JhcGggbnVtYmVyLiBUaGUgd2hvbGUtZ3JhcGggZGV2aWNlCiAgICAgICAgdGltZSBpcyB0aGVyZWZvcmUgbWVhc3VyZWQgYXMgYHN1bSBvZiBwZXIta2VybmVsIGRldmljZSB0aW1lc2AgYnkKICAgICAgICB3YWxraW5nIHRoZSBMSU5FQVIsIHdoaWNoIGBsaW5lYXJfZGV2aWNlX3RpbWVfbXNgIGRvZXMuCgogICAgUmV0dXJucyBOb25lIHdoZW4gdGhlIGJhY2tlbmQgZG9lcyBub3Qgc3VwcG9ydCBpdCwgcmF0aGVyIHRoYW4gYSB6ZXJvIHRoYXQKICAgIHdvdWxkIHJlYWQgYXMgImluZmluaXRlbHkgZmFzdCIuCiAgICAiIiIKICAgIGZyb20gdGlueWdyYWQuZW5naW5lLnJlYWxpemUgaW1wb3J0IHRpbWVfY2FsbAogICAgdHJ5OgogICAgICAgIGxpbmVhciwgdmFyX3ZhbHMgPSBfbGluZWFyX2FuZF92YXJzKHRlbnNvcikKICAgIGV4Y2VwdCBFeGNlcHRpb24gYXMgZToKICAgICAgICByZXR1cm4geyJlcnJvciI6IGYie3R5cGUoZSkuX19uYW1lX199OiB7ZX0iWzoyMDBdfQogICAgdHJ5OgogICAgICAgIGJlc3QgPSBOb25lCiAgICAgICAgZm9yIGNhbGwgaW4gbGluZWFyLnNyYzoKICAgICAgICAgICAgdCA9IG5leHQodGltZV9jYWxsKGNhbGwsIHZhcl92YWxzPXZhcl92YWxzKSkKICAgICAgICAgICAgYmVzdCA9IHQgaWYgYmVzdCBpcyBOb25lIGVsc2UgbWF4KGJlc3QsIHQpCiAgICAgICAgcmV0dXJuIHsibWF4X2tlcm5lbF9tcyI6IGJlc3QgKiAxZTMgaWYgYmVzdCBpcyBub3QgTm9uZSBlbHNlIE5vbmV9CiAgICBleGNlcHQgRXhjZXB0aW9uIGFzIGU6CiAgICAgICAgcmV0dXJuIHsiZXJyb3IiOiBmInt0eXBlKGUpLl9fbmFtZV9ffToge2V9Ils6MjAwXX0KCgpkZWYgX2xpbmVhcl9hbmRfdmFycyh0ZW5zb3IpOgogICAgZnJvbSB0aW55Z3JhZCBpbXBvcnQgVGVuc29yIGFzIF9UCiAgICBsaW4sIHZ2ID0gX1QubGluZWFyX3dpdGhfdmFycyh0ZW5zb3IpCiAgICByZXR1cm4gbGluLCB2dgoKCmRlZiBsaW5lYXJfZGV2aWNlX3RpbWVfbXModGVuc29yLCBpdGVyczogaW50ID0gMykgLT4gZGljdCB8IE5vbmU6CiAgICAiIiJTdW0gb2YgcGVyLWtlcm5lbCBkZXZpY2UgdGltZXMgb3ZlciB0aGUgd2hvbGUgZ3JhcGgsIGluIG1zLgoKICAgIFdhbGtzIHRoZSBjb21waWxlZCBMSU5FQVIgYW5kIHRpbWVzIGVhY2ggQ0FMTCB3aXRoIGB0aW1lX2NhbGxgLCB0aGVuIHN1bXMuCiAgICBUaGlzIGlzIHRoZSB3aG9sZS1ncmFwaCBkZXZpY2UgdGltZTogaXQgaXMgd2hhdCB0aGUgcmVmZXJlbmNlJ3MKICAgIHdob2xlLXN0ZXAgbnVtYmVyIGNvbnRhaW5zLCBvbiB0aGUgc2FtZSBraW5kIG9mIGNsb2NrLCBhbmQgaXQgZXhjbHVkZXMgaG9zdAogICAgdHJhY2luZywgd2hpY2ggdGhlIGZyYW1ld29yayBhZGRzIGFuZCB0aGUgcmVmZXJlbmNlIGRvZXMgbm90LgogICAgIiIiCiAgICBmcm9tIHRpbnlncmFkLmVuZ2luZS5yZWFsaXplIGltcG9ydCB0aW1lX2NhbGwKICAgIHRyeToKICAgICAgICBsaW5lYXIsIHZhcl92YWxzID0gX2xpbmVhcl9hbmRfdmFycyh0ZW5zb3IpCiAgICBleGNlcHQgRXhjZXB0aW9uIGFzIGU6CiAgICAgICAgcmV0dXJuIHsiZXJyb3IiOiBmInt0eXBlKGUpLl9fbmFtZV9ffToge2V9Ils6MjAwXX0KICAgIHRyeToKICAgICAgICBjYWxscyA9IGxpc3QobGluZWFyLnNyYykKICAgICAgICBmb3IgXyBpbiByYW5nZSgyKTogICAgICAgICAgICAgICAgICAgICAjIHR3byB3YXJtIHBhc3NlcwogICAgICAgICAgICBmb3IgYyBpbiBjYWxsczoKICAgICAgICAgICAgICAgIG5leHQodGltZV9jYWxsKGMsIHZhcl92YWxzPXZhcl92YWxzKSkKICAgICAgICB0b3QgPSBbXQogICAgICAgIGZvciBfIGluIHJhbmdlKG1heCgxLCBpdGVycykpOgogICAgICAgICAgICBzID0gMC4wCiAgICAgICAgICAgIGZvciBjIGluIGNhbGxzOgogICAgICAgICAgICAgICAgcyArPSBuZXh0KHRpbWVfY2FsbChjLCB2YXJfdmFscz12YXJfdmFscykpCiAgICAgICAgICAgIHRvdC5hcHBlbmQocyAqIDFlMykKICAgICAgICB0b3Quc29ydCgpCiAgICAgICAgcmV0dXJuIHsibl9rZXJuZWxzIjogbGVuKGNhbGxzKSwgIm1lZGlhbl9tcyI6IHN0YXRpc3RpY3MubWVkaWFuKHRvdCksCiAgICAgICAgICAgICAgICAibWluX21zIjogdG90WzBdLCAibWF4X21zIjogdG90Wy0xXSwKICAgICAgICAgICAgICAgICJtZWRpYW5fcGVyX2tlcm5lbF91cyI6IHN0YXRpc3RpY3MubWVkaWFuKHRvdCkgKiAxZTMgLyBtYXgoMSwgbGVuKGNhbGxzKSl9CiAgICBleGNlcHQgRXhjZXB0aW9uIGFzIGU6CiAgICAgICAgcmV0dXJuIHsiZXJyb3IiOiBmInt0eXBlKGUpLl9fbmFtZV9ffToge2V9Ils6MjAwXX0KCgpkZWYgYmVuY2hfcmVuZGVyKGJ1aWxkX2lucHV0cywgcmVuZGVyX2ZuLCBkZXYsIGl0ZXJzOiBpbnQsIGxhYmVsOiBzdHIsCiAgICAgICAgICAgICAgICAgd2FybXVwOiBpbnQgPSA1KSAtPiBkaWN0OgogICAgIiIiQnVpbGQgb25jZSwgdGltZSB0aGUgcmVhbGl6ZSwgYW5kIHJlcG9ydCB0aGUgYWxsb2NhdG9yJ3MgcGVhay4KCiAgICBgYnVpbGRfaW5wdXRzKClgIG11c3QgcmV0dXJuIHdoYXRldmVyIGByZW5kZXJfZm5gIG5lZWRzLiBJdCBpcyBjYWxsZWQgaW5zaWRlCiAgICB0aGUgbWVtb3J5IHByb2JlLCBzbyB0aGUgcmVwb3J0ZWQgcGVhayBjb3ZlcnMgdGhlIGlucHV0cyAqYW5kKiBldmVyeQogICAgaW50ZXJtZWRpYXRlIHRoZSByZW5kZXIgYWxsb2NhdGVzIC0tIHdoaWNoIGlzIHRoZSByaWdodCBib3VuZGFyeSwgYmVjYXVzZQogICAgdGhlIHJlZmVyZW5jZSdzIHBlci1rZXJuZWwgbWVhc3VyZW1lbnQgYWxzbyBydW5zIGFnYWluc3QgYSBmdWxseSBhbGxvY2F0ZWQKICAgIGVudmlyb25tZW50IChgYmVuY2guY3VgIGFsbG9jYXRlcyB0aGUgd2hvbGUgc3RhdGUgdXAgZnJvbnQgYW5kIHRpbWVzIGEKICAgIHNpbmdsZSBsYXVuY2ggaW5zaWRlIGl0KS4KICAgICIiIgogICAgb3V0OiBkaWN0ID0geyJsYWJlbCI6IGxhYmVsLCAicmVhbGl6ZWQiOiBGYWxzZSwgImVycm9yIjogTm9uZX0KICAgIGdjLmNvbGxlY3QoKQogICAgciA9IE5vbmUKICAgIHdpdGggTWVtUHJvYmUoZGV2KSBhcyBtcDoKICAgICAgICB0cnk6CiAgICAgICAgICAgIGlucHV0cyA9IGJ1aWxkX2lucHV0cygpCiAgICAgICAgICAgIG91dFsibWVtX2FmdGVyX2lucHV0cyJdID0gbXAuYXNfZGljdCgpCiAgICAgICAgICAgIG1wLnJlc2V0KCkKICAgICAgICAgICAgciA9IHJlbmRlcl9mbigqaW5wdXRzKSBpZiBpc2luc3RhbmNlKGlucHV0cywgdHVwbGUpIGVsc2UgcmVuZGVyX2ZuKGlucHV0cykKICAgICAgICAgICAgciA9IHJbMF0gaWYgaXNpbnN0YW5jZShyLCB0dXBsZSkgZWxzZSByCiAgICAgICAgICAgICMgZGV2aWNlIHRpbWluZyBGSVJTVCwgd2hpbGUgdGhlIGdyYXBoIGlzIHN0aWxsIHVucmVhbGl6ZWQ6CiAgICAgICAgICAgICMgYGxpbmVhcl93aXRoX3ZhcnNgIG9uIGEgcmVhbGl6ZWQgdGVuc29yIHJldHVybnMgYW4gZW1wdHkgTElORUFSLAogICAgICAgICAgICAjIGJlY2F1c2UgdGhlcmUgaXMgbm90aGluZyBsZWZ0IHRvIHJ1bi4KICAgICAgICAgICAgb3V0WyJkZXZpY2VfdGltZSJdID0gbGluZWFyX2RldmljZV90aW1lX21zKAogICAgICAgICAgICAgICAgciwgaXRlcnM9bWluKDUsIG1heCgyLCBpdGVycyAvLyAyKSkpCiAgICAgICAgICAgIHIucmVhbGl6ZSgpCiAgICAgICAgICAgIG91dFsicmVhbGl6ZWQiXSA9IFRydWUKICAgICAgICBleGNlcHQgRXhjZXB0aW9uIGFzIGU6ICAgICAgICAgICAgICAgICAgIyBPT00gaXMgYSByZXN1bHQsIG5vdCBhIGNyYXNoCiAgICAgICAgICAgIG91dFsiZXJyb3IiXSA9IGYie3R5cGUoZSkuX19uYW1lX199OiB7c3RyKGUpfSJbOjQwMF0KICAgICAgICBfc3luYyhkZXYpCiAgICAgICAgb3V0WyJtZW0iXSA9IG1wLmFzX2RpY3QoKQoKICAgIGlmIG5vdCBvdXRbInJlYWxpemVkIl06CiAgICAgICAgcmV0dXJuIG91dAoKICAgIG91dFsib2JzX2J5dGVzIl0gPSBpbnQoci5uYnl0ZXMoKSkKICAgIG91dFsib2JzX3NoYXBlIl0gPSBsaXN0KHIuc2hhcGUpCiAgICBvdXRbIm9ic19kdHlwZSJdID0gc3RyKHIuZHR5cGUpCiAgICByZXR1cm4gb3V0CgoKZGVmIGQyaF90aW1lKHIsIGRldiwgaXRlcnM6IGludCwgd2FybXVwOiBpbnQgPSAzKSAtPiBkaWN0OgogICAgIiIiVGhlIGNvc3Qgb2YgcmVhZGluZyB0aGUgb2JzZXJ2YXRpb24gYmFjaywgYXMgaXRzIG93biBudW1iZXIuIiIiCiAgICByZXR1cm4gdGltZV9jYWxsKGxhbWJkYTogci5udW1weSgpLCBkZXYsIGl0ZXJzLCB3YXJtdXApCg==", "7ec6927c711e0eec2b4de27d8eb03bb6c2f7e0d660edd8dfa969dba61434c539"], ["minecu_tg_faithful/__init__.py", "IiIibWluZWN1X3RnX2ZhaXRoZnVsIC0tIHRoZSByZWZlcmVuY2UncyBzdHJhdGVneSBpbiBUaW55R3JhZCwgY3Jvc3NpbmdzIG1hcmtlZC4KClRoaXMgYXJtIGFuc3dlcnMgYSBkaWZmZXJlbnQgcXVlc3Rpb24gZnJvbSB0aGUgbmFpdmUgcG9ydCAoYG1pbmVjdV90Zy9gKSBhbmQKdGhlIGNvbXBldGVudCBwb3J0IChgbWluZWN1X3RnX29wdC9gKS4gVGhvc2UgdHdvIGNvbXBhcmUgYSBoYW5kLXdyaXR0ZW4ga2VybmVsCmFnYWluc3Qgd2hhdCBUaW55R3JhZCBnZW5lcmF0ZXMgZnJvbSAqaXRzIG93biogaWRlYXMsIHdoaWNoIGNhbm5vdCBzZXBhcmF0ZQoidGhlIGNvbXBpbGVyIGZhaWxlZCB0byBmaW5kIHNvbWV0aGluZyBhIGNvc3QgbW9kZWwgY291bGQgZmluZCIgZnJvbSAidGhlCm9wdGltaXNhdGlvbiBsaXZlcyBiZWxvdyB0aGUgYWJzdHJhY3Rpb24iLgoKVGhpcyBvbmUgcmVwcm9kdWNlcyB0aGUgcmVmZXJlbmNlJ3MgYWN0dWFsIHN0cmF0ZWd5LCBpbmNsdWRpbmcgdGhlIHBhcnRzIGJlbG93CnRoZSBhYnN0cmFjdGlvbiwgYW5kIHJlY29yZHMgZXZlcnkgcGxhY2UgaXQgaGFzIHRvIGNyb3NzIHRvIGdldCB0aGVyZS4gV2hlcmUKdGhlIGJvdW5kYXJ5IHR1cm5zIG91dCB0byBiZSAqaXMgdGhlIHJlc3VsdCouCgpgVGVuc29yLmN1c3RvbV9rZXJuZWxgIGlzIG5vdCB1c2VkLiBObyBgLmN1YCBpcyB3cml0dGVuIG9yIGNvbXBpbGVkIGFsb25nc2lkZS4KRXZlcnl0aGluZyB0aGF0IHJ1bnMgaXMgQ1VEQSB0aGF0IFRpbnlHcmFkJ3Mgb3duIHJlbmRlcmVyIGVtaXR0ZWQuCgogICAgcmVuZGVyLnB5ICAgICAgIHRoZSBwb3J0LiBgbW9kZT0iY2xlYW4iYCBhbmQgYG1vZGU9ImZhaXRoZnVsImAuCiAgICBjcm9zc2luZ3MucHkgICAgZXZlcnkgY3Jvc3NpbmcsIHdpdGggVGlueUdyYWQgYGZpbGU6bGluZWAsIHBsdXMgdGhlIGdhcHMuCiAgICB2ZXJpZnkucHkgICAgICAgdGhlIGNvcnJlY3RuZXNzIGdhdGU6IGJvdGggbW9kZXMsIGFnYWluc3QgdGhlIHJlZmVyZW5jZSdzCiAgICAgICAgICAgICAgICAgICAgb3duIENQVSBtb2RlbCBhbmQgYW4gaW5kZXBlbmRlbnQgZmxvYXQ2NCB0cmFuc2NyaXB0aW9uLgogICAgY2Vuc3VzLnB5ICAgICAgIGxhdW5jaCBhbmQgcHJvZ3JhbSBjb3VudHMsIG9mZmxpbmUsIGZvciBib3RoIG1vZGVzLgogICAgcHJvYmVfa25vYnMucHkgIHdoaWNoIENvbnRleHRWYXJzIHJlYWNoIHRoaXMgZ3JhcGggYXQgYWxsLgogICAgc2Fzcy5weSAgICAgICAgIHRoZSBTQVNTIGNlbnN1cywgYm90aCBzaWRlcywgdGhyb3VnaCBvbmUgY2F0ZWdvcmlzZXIuCiAgICBrYWdnbGUvICAgICAgICAgdGhlIFQ0IHB1c2g6IGJvdGggbW9kZXMgYW5kIHRoZSBoYW5kLXdyaXR0ZW4gYmFzZWxpbmUsCiAgICAgICAgICAgICAgICAgICAgb25lIGNhcmQsIG9uZSBydW4uCgpSZWFkIGByZW5kZXIucHlgJ3MgbW9kdWxlIGRvY3N0cmluZyBhbmQgdGhlbiBgY3Jvc3NpbmdzLnB5YCBiZWZvcmUgcXVvdGluZwphbnl0aGluZyBhYm91dCB3aGF0IHRoaXMgY29kZSBpcy4KIiIiCgpmcm9tIC5yZW5kZXIgaW1wb3J0IGNhbWVyYV9iYXNpcywgcmVuZGVyCgpfX2FsbF9fID0gWyJjYW1lcmFfYmFzaXMiLCAicmVuZGVyIl0K", "cecf6f98be10403115d4db675ad31d2061d79dcab01370e607467ecfe288ef6d"], ["minecu_tg_faithful/render.py", "", "85b5a0b5fb843e9791b3ef8303a88724483be51544ee18af795f200ffe152419"], ["minecu_tg_faithful/crossings.py", "", "c0d94bc0c822ccc490e5dbb185622aa9f37edd086f7786a56ba4c4b8c94e4e5d"], ["minecu_tg_faithful/verify.py", "", "111b32e7a094d6456e1dbf08b81111bf4255487624bdb09da939c16f27173917"], ["minecu_tg_faithful/battery.py", "", "d5b92e7eccf077deb7db8f70081ad9f354fb0c3dc28cff41d1121102815f7c5d"], ["minecu_ref/bench/dda_model.py", "", "f5ff7fe2e15f7d9b1313484187747da987c0d5fc165ec35b30daf3b10fc049bd"], ["minecu_ref/src/kernels.cu", "", "c6ed8f06ee64d40a8a8ab7ebce3b3ceea1e3ac93f9ee50cfc1e63d3f090d878f"], ["minecu_tg/kaggle/refdrv.cu", "", "d3a0931d642d67bd5e6fbe22e77a732da72edc3502ca8296f87f615a224965e8"]]
BATCHES = [512, 1024, 2048]
ITERS = 10
WARMUP = 3
REFDRV_LIMIT_S = 300
GATE_LIMIT_S = 900
BATTERY_LIMIT_S = 2400

lines = []
# Write-through, so a run the platform kills still leaves its evidence on the
# card. `kaggle kernels output` fetches /kaggle/working, and a summary that
# only exists after the last stage exists for exactly the runs that did not
# reach the last stage -- which are the runs anybody needs to read. The cost is
# one small file write per printed line.
SUMMARY = f"{OUT}/SUMMARY.txt"
def say(*a):
    s = " ".join(str(x) for x in a)
    print(s, flush=True)
    lines.append(s)
    try:
        open(SUMMARY, "w").write("\n".join(lines) + "\n")
    except OSError:
        pass

def die(code, why):
    say("")
    say("FATAL: " + why)
    lines.append(f"FATAL {why}")
    try:
        open(SUMMARY, "w").write("\n".join(lines) + "\n")
    except OSError:
        pass
    sys.exit(code)

def sh(cmd, **kw):
    p = subprocess.run(cmd, shell=isinstance(cmd, str), capture_output=True,
                       text=True, **kw)
    return p.returncode, p.stdout, p.stderr

def stream(cmd, label, limit=None, **kw):
    """Run a long subprocess, echoing its output as it appears.

    `sh` captures everything and says nothing until the child exits, which for
    the on-card gate and the battery is minutes of silence. A silent notebook is
    indistinguishable from a hung one: it is why "it hung" was the first
    hypothesis for the 50-second kill, and why the 2136 s block in
    `out/minecu_tg/20260928-221859/` is unreadable. This echoes line by line.

    `limit` is seconds, and it is enforced HERE rather than left to the
    platform's run cap. Echoing without a bound would replace "the harness
    timed this stage out" with "the session was killed at an arbitrary point",
    which is the failure this notebook already had twice; a stage that overruns
    says so, by name, with its last 60 lines.
    """
    import time as _t
    t0 = _t.time()
    say(f"  [{label}] starting"
        + (f", bounded at {limit:.0f}s" if limit else ", unbounded"))
    p = subprocess.Popen(cmd, stdout=subprocess.PIPE, stderr=subprocess.STDOUT,
                         text=True, bufsize=1, **kw)
    tail = []
    killed = False
    for line in p.stdout:
        line = line.rstrip()
        tail.append(line)
        del tail[:-60]
        say(f"  [{label} {_t.time() - t0:6.1f}s] {line}")
        if limit and _t.time() - t0 > limit and p.poll() is None:
            say(f"  [{label}] OVER its {limit:.0f}s bound after "
                f"{_t.time() - t0:.0f}s; killing it and reporting what it did")
            p.kill()
            killed = True
            break
    rc = p.wait()
    say(f"  [{label}] exit {rc} after {_t.time() - t0:.1f}s"
        + ("  (killed by the bound)" if killed else ""))
    return (124 if killed else rc), "\n".join(tail), None

def dump_summary():
    open(SUMMARY, "w").write("\n".join(lines) + "\n")

say(f"  session budget: gate <= {GATE_LIMIT_S}s, battery <= {BATTERY_LIMIT_S}s, "
    f"refdrv <= {REFDRV_LIMIT_S}s. The platform cap is set in run.sh and these "
    f"must fit inside it.")
say("=" * 78)
say("minecu_tg_faithful -- the reference's strategy in TinyGrad, both modes,")
say("against the hand-written CUDA render, on this T4")
say("=" * 78)

# ----------------------------------------------------------- 1. provenance
say("")
say("--- 1. provenance: every payload hash-checked before anything is built ---")
for name, b64, want in PAYLOADS:
    data = base64.b64decode(b64)
    got = hashlib.sha256(data).hexdigest()
    if got != want:
        die(9, f"{name} sha256 {got} != {want}; refusing to measure a source I "
               f"cannot identify. Regenerate with minecu_tg_faithful/kaggle/run.sh.")
    dest = f"{OUT}/{name}"                             # repo-relative: the
    os.makedirs(os.path.dirname(dest), exist_ok=True) # package stays a package
    open(dest, "wb").write(data)
    say(f"  {name:38s} {len(data):8d} B  sha256 {got[:16]}  OK")
sys.path.insert(0, OUT)

# --------------------------------------------------------------- 2. tinygrad
say("")
say("--- 2. tinygrad ---")
rc, out, err = sh([sys.executable, "-m", "pip", "install", "-q", "tinygrad"])
say(f"  pip install tinygrad rc={rc}")
if rc != 0:
    say("  " + (err or out)[-400:].replace("\n", "\n  "))
try:
    import tinygrad
    from tinygrad.helpers import VERSION
    say(f"  tinygrad at {tinygrad.__file__}")
    say(f"  helpers.VERSION (the program-cache schema) = {VERSION}")
except Exception as e:
    die(12, f"tinygrad did not import: {e}")

# ONLY NOW, with tinygrad importable, check that what we laid down imports.
# This has to come after the install or it fails on a missing tinygrad instead.
try:
    import importlib
    for m in ("minecu_tg.render", "minecu_tg.cpu_reference", "minecu_tg.bench",
              "minecu_tg_faithful.render", "minecu_tg_faithful.crossings"):
        importlib.import_module(m)
        say(f"  import {m:34s} OK")
except Exception as e:
    die(13, f"payload does not import: {type(e).__name__}: {e}")

# ------------------------------------------------------------------ 3. the
# card. verify.py runs again here, not just on the host: it proves the same
# source is correct against the card's tinygrad, which may not be this host's.
say("")
say("--- 3. correctness gate ON THE CARD ---")
# DEV=CUDA, not the default CPU. Two reasons, and the first is the one that
# matters: it verifies the render on the device every timing below is taken
# on, so a codegen difference on the card cannot survive to become a wrong
# image in the summary. The second is that this image has no `clang` and
# tinygrad's CPU backend shells out to one, so DEV=CPU dies with
# FileNotFoundError before the first check runs.
rc, out, err = stream([sys.executable, f"{OUT}/minecu_tg_faithful/verify.py",
                       "--quick", "--dev", "CUDA"],
                      "gate", limit=GATE_LIMIT_S, cwd=OUT,
                      env={**os.environ, "VERIFY_LOG": f"{OUT}/verify_card.log"})
if rc != 0:
    say("  " + ((err or out) or "")[-1500:].replace("\n", "\n  "))
    die(14, "the correctness gate failed on the card; no timing is reported "
            "because a timing from a wrong render is a measurement of the "
            "wrong thing")
say("  PASS on the card itself. The observation is bit-identical to the naive")
say("  port and to an independent float64 transcription, and both modes agree")
say("  on the per-pixel DDA step count.")

# ------------------------------------------------------------- 4. device
say("")
say("--- 4. device ---")
import subprocess as _sp
try:
    smi = _sp.run(["nvidia-smi",
                   "--query-gpu=name,memory.total,compute_cap,driver_version",
                   "--format=csv,noheader"], capture_output=True, text=True)
    say("  " + smi.stdout.strip())
except Exception as e:
    say(f"  nvidia-smi unavailable: {e}")
from tinygrad import Device
say(f"  DEV={Device.DEFAULT}")

# ------------------------------------------------- 5. build the baseline
# refdrv.cu `#include`s kernels.cu (refdrv.cu:36-38, via
# `#ifndef MINECU_KERNELS`), so it is compiled ALONE -- passing kernels.cu on
# the command line as well compiles it a second time and the LINKER reports
# `multiple definition of launch_render_uint8_prebasis' and eight others.
#
# And `#include` resolves relative to the INCLUDING file, which is
# `/kaggle/working/minecu_tg/kaggle/refdrv.cu`, while kernels.cu is written to
# `/kaggle/working/minecu_ref/src/kernels.cu`. So the default include finds
# nothing: push 7 died here with
#     refdrv.cu:36:24: fatal error: kernels.cu: No such file or directory
# which is a compile error in a file that is correct, about a path this stage
# owns. `-DMINECU_KERNELS="<abs path>"` is the override refdrv.cu documents for
# exactly this (`minecu_tg/kaggle/make_notebook.py:210` gets it right).
#
# A pre-flight that only compiled refdrv.cu with the wrong flags would pass
# here and fail in the notebook, which is why this stage exists separately.
say("")
say("--- 5. build the hand-written side for sm_75 ---")
NVCC = None
for _p in ("/usr/local/cuda/bin/nvcc", "nvcc"):
    if sh([_p, "--version"])[0] == 0:
        NVCC = _p
        break
if NVCC is None:
    die(10, "no nvcc on this runner; the T4 image is supposed to have one")
say("  " + "\n  ".join(sh([NVCC, "--version"])[1].strip().splitlines()[-2:]))
os.makedirs(f"{OUT}/ref", exist_ok=True)
_cmd = [NVCC, "-O3", "--use_fast_math", "-std=c++17", "-arch=sm_75",
        "-diag-suppress", "177",
        f'-DMINECU_KERNELS="{OUT}/minecu_ref/src/kernels.cu"',
        f"{OUT}/minecu_tg/kaggle/refdrv.cu", "-o", f"{OUT}/ref/refdrv"]
say("  " + " ".join(_cmd))
rc, out, err = stream(_cmd, "nvcc", limit=REFDRV_LIMIT_S)
if rc != 0:
    say("  " + (err or out)[-1500:].replace("\n", "\n  "))
    die(20, "refdrv did not build; the hand-written side is required, because "
            "a comparison with only one side is not a comparison")
say("  OK  {0} bytes".format(os.path.getsize(f"{OUT}/ref/refdrv")))

# ---------------------------------------------------------- 6. the timings
say("")
say("--- 6. the hand-written baseline, per batch (cudaEvent, its own clock) ---")
ref_render_us = {}
for b in BATCHES:
    os.makedirs(f"{OUT}/refdump/b{b}", exist_ok=True)
    cmd = [f"{OUT}/ref/refdrv", "--batch", str(b), "--world", "16", "--res", "32",
           "--max-steps", "32", "--ground", "4", "--iters", str(ITERS),
           "--warmup", str(WARMUP), "--outdir", f"{OUT}/refdump/b{b}", "--dump"]
    rc, out, err = stream(cmd, f"refdrv b{b}", limit=REFDRV_LIMIT_S)
    if rc != 0:
        say(f"  batch {b:6d}  refdrv FAILED rc={rc}")
        say("  " + ((err or out) or "")[-800:].replace("\n", "\n  "))
        continue
    said_step = False
    for line in out.splitlines():
        ls = line.strip()
        if ls.startswith("RENDER_US"):
            p = ls.split()
            ref_render_us[b] = float(p[1])
            say(f"  batch {b:6d}  RENDER_US median {p[1]:>10s}  min {p[2]:>10s}  "
                f"max {p[3]:>10s}  iters {p[4]}")
        elif ls.startswith("STEP_US"):
            if not said_step:
                said_step = True
                say(f"  batch {b:6d}  STEP_US   {ls.split()[1]:>10s}   "
                    f"<- CONTEXT ONLY. A whole step is 7 launches, five of them")
                say(f"                          state updates. It is never a "
                    f"denominator here.")
        elif "steps/pixel" in ls or "DDA" in ls:
            say(f"  batch {b:6d}  {ls}")
json.dump(ref_render_us, open(f"{OUT}/ref_render_us.json", "w"))

# -------------------------------------------------------- 7. both modes
say("")
say("--- 7. this port, both modes, same batches, same run, same process ---")
rc, out, err = stream([sys.executable, f"{OUT}/minecu_tg_faithful/battery.py",
                       "--batches", ",".join(str(b) for b in BATCHES),
                       "--refdrv", f"{OUT}/ref/refdrv",
                       "--ref-outdir", f"{OUT}/refdump",
                       "--json", f"{OUT}/faithful_battery.json"],
                      "battery", limit=BATTERY_LIMIT_S, cwd=OUT)
if rc != 0:
    die(30, f"battery.py exited {rc}")

# --------------------------------------------------------- 8. the summary
say("")
say("--- 8. the three-way table, and the clean-vs-faithful delta ---")
try:
    rep = json.load(open(f"{OUT}/faithful_battery.json"))
except Exception as e:
    die(31, f"no battery json: {e}")
say(f"  {'batch':>6}  {'ref us':>10}  {'clean ms':>10}  {'faith ms':>10}  "
    f"{'f/clean':>8}  {'clean/ref':>9}  {'faith/ref':>9}  {'kernels c/f':>12}  same obs")
say(f"  {'(DEVICE time; host tracing excluded -- see battery section 5)':>70}")
for row in rep.get("rows", []):
    def g(k, w=10, p=3):
        v = row.get(k)
        return f"{v:{w}.{p}f}" if isinstance(v, (int, float)) else f"{'?':>{w}}"
    b = row["batch"]
    rr = ref_render_us.get(b, row.get("ref_render_us"))
    say(f"  {b:>6}  {rr if rr else float('nan'):>10.1f}  {g('clean_ms'):>10}  "
        f"{g('faithful_ms'):>10}  {g('faithful_over_clean', 8, 3):>8}  "
        f"{g('clean_over_ref', 9, 2):>9}  {g('faithful_over_ref', 9, 2):>9}  "
        f"{str(row.get('clean_kernels','?')) + '/' + str(row.get('faithful_kernels','?')):>12}  "
        f"{row.get('obs_identical_modes')}")

say("")
say("--- what these numbers are, and are not ---")
say("  * The three columns are RENDERS. Same world, same cameras, same")
say("    max_steps, same uint8 observation, same card, same run, equal batch.")
say("    `f/clean` is the crossing cost and is the number this arm adds; the")
say("    two `*/ref` columns are render-against-render and are the same")
say("    comparison `notes/19` §5.2 makes.")
say("  * The two ms columns are DEVICE time -- the sum of the per-kernel")
say("    device times in the graph, measured with tinygrad's own time_call,")
say("    host tracing excluded. The reference's us figure is a cudaEvent")
say("    median around a single launch, which is also device time and is the")
say("    like-for-like comparison. The host build+realize figure printed")
say("    beside each row is CONTEXT: it includes the framework's graph")
say("    tracing, which the reference does not pay, so it is NOT divided by")
say("    anything here.")
say("  * Calling realize() on an already-realized tensor is a silent no-op.")
say("    An earlier version of battery.py timed exactly that and reported the")
say("    port at 0.011 ms against the reference's 152 us -- a 13x speedup over")
say("    hand-written CUDA, from a graph whose offline SASS is tens of times")
say("    the reference's instruction count. time_graph now raises if handed a")
say("    realized tensor, and the timing instrument is minecu_tg's own")
say("    linear_device_time_ms, which is the code path `notes/19` measured")
say("    with. That false number is the reason the clock is named on every")
say("    row rather than in a footnote.")
say("  * The reference's whole-step figure is printed for context and is NOT a")
say("    denominator anywhere. `4.55 M env-steps/s` covers 7 launches.")
say("  * `same obs` compares the two modes' observations by sha256. Both are")
say("    verified against the same float64 transcription and the same naive")
say("    port, so a False here would mean the crossings changed the")
say("    computation, and the ratios above it would be between two different")
say("    computations.")
say("  * The ratio to the reference is NOT attributed to the compiler unless")
say("    both sides did the same work. Each mode's mean steps/pixel is in the")
say("    battery output. The reference's own DEVICE-side step count is NOT")
say("    available here: it needs the dda_probe_kernel from")
say("    minecu_ref/bench/bench.cu, which refdrv.cu does not link, and putting")
say("    the CPU model's number under the same label would be a claim about a")
say("    device that was not measured. The workload check is one-sided and")
say("    the note says so rather than rounding it up.")
say("  * The obs_sha256 in the battery is this port's. verify.py proved it")
say("    equals the naive port's and a float64 transcription's ON THIS CARD in")
say("    this run. It is NOT compared against the reference's dumped bytes")
say("    here; that is notes/19's comparison on the reference's own state, and")
say("    this arm did not re-derive it.")
say("  * No SASS number is reported here. The instruction and register counts")
say("    come from the offline archlab route on the host, where ptxas and")
say("    cuobjdump exist and nvcc's version is pinned; the T4 image's CUDA")
say("    version is recorded but not used for a count.")

dump_summary()
say("")
say("wrote /kaggle/working/SUMMARY.txt")
```
